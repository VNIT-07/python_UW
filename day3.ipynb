### Operators like arithmatic, logical.. type casting in detail

### I. Operators:

- Arithmetic operators

- Assignment operators

- Comparison operators

- Logical operators

- Identity operators

- Membership operators

- Bitwise operators


```python
# 1. Arithmetic Operations

a = 15
b = 4

print("Addition:", a + b)  

print("Subtraction:", a - b) 

print("Multiplication:", a * b)  

print("Division:", a / b) 

print("Floor Division:", a // b)  

print("Modulus:", a % b) 

print("Exponentiation:", a ** b)
```

```python
# 2. Assignment Operators

a = 10
b = a
print(b)
b += a
print(b)
b -= a
print(b)
b *= a
print(b)
b <<= a
print(b)
```

```python
# 3. Comparison Operators

a = 13
b = 33

print(a > b)
print(a < b)
print(a == b)
print(a != b)
print(a >= b)
print(a <= b)
```

```python
# 4. Logical Operators

a = True
b = False
print(a and b)
print(a or b)
print(not a)
```

```python
# 5. Identity Operators

a = 5
b = 5
print(a is b)
print(a is not b)
```

```python
# 6. Membership Operators

x = 24
y = 20
my_list = [10, 20, 30, 40, 50]

if (x not in my_list):
    print("x is NOT present in given list")
else:
    print("x is present in given list")

if (y in my_list):
    print("y is present in given list")
else:
    print("y is NOT present in given list")
```

```python
# 7. Bitwise Operators

a = 10
b = 4

print(a & b)
print(a | b)
print(~a)
print(a ^ b)
print(a >> 2)
print(a << 2)
```

### III. Type Casting

1. Implicit Casting

```python
# Python automatically converts 'a' to int 
a = 7
print(type(a)) 

# Python automatically converts 'b' to float 
b = 3.0
print(type(b)) 

# Python automatically converts 'c' to float as it is a float addition 
c = a + b 
print(c, type(c))

# Python automatically converts 'd' to float as it is a float multiplication
d = a * b
print(d, type(d))
# print()
```

2. Explicit Casting

```python
# Converting Int to Float in with the float() function.

a = 5
n = float(a)

print(n)
print(type(n))
```

```python
# Converting Float to int datatype in Python with int() function.

a = 5
# typecast to str
n = str(a)

print(n)
print(type(n))
```

```python
# Casting string data type into float data type with float() function.

a = "5.9"
n = float(a)

print(n)
print(type(n))
```

```python
# Converting string to int datatype in Python with int() function. If the given string is not number, then it will throw an error.

a = "5"
b = 't'
n = int(a)

print(n)
print(type(n))

# print(int(b))
print(type(b))
```

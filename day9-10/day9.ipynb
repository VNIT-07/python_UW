

# Python Chapter 9 – Control Flow

Control flow determines the order in which a program executes statements.

This chapter covers:

- Conditional statements (`if`, `elif`, `else`)
- `for` loops
- `while` loops
- Loop control statements (`break`, `continue`, `pass`)
- Comprehension syntax

---

# 1. Conditional Statements

Python uses conditional statements to execute code based on a condition.

## if

Execute code only when a condition is `True`.

```python
age = 20

if age >= 18:
    print("Adult")
```

---

## if...else

Choose between two blocks of code.

```python
age = 15

if age >= 18:
    print("Adult")
else:
    print("Minor")
```

---

## if...elif...else

Check multiple conditions.

```python
marks = 85

if marks >= 90:
    print("A")
elif marks >= 75:
    print("B")
elif marks >= 60:
    print("C")
else:
    print("Fail")
```

Python evaluates conditions from top to bottom.

Only the first matching condition executes.

---

## Nested if

An `if` statement may contain another `if`.

```python
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
```

---

## Comparison Operators

| Operator | Meaning |
|----------|---------|
| == | Equal |
| != | Not equal |
| > | Greater than |
| < | Less than |
| >= | Greater than or equal |
| <= | Less than or equal |

---

## Logical Operators

### and

Both conditions must be true.

```python
if age >= 18 and has_id:
    ...
```

### or

At least one condition must be true.

```python
if is_admin or is_owner:
    ...
```

### not

Reverses a boolean value.

```python
if not logged_in:
    ...
```

---

## Truthy and Falsy Values

Falsy values include:

- False
- None
- 0
- ""
- []
- {}
- ()
- set()

Everything else is considered truthy.

Example:

```python
users = []

if not users:
    print("No users found")
```

# 2. for Loops

A `for` loop iterates over an iterable.

```python
for number in range(5):
    print(number)
```

Output:

```
0
1
2
3
4
```

---

## Iterating over Lists

```python
names = ["Alice", "Bob", "Charlie"]

for name in names:
    print(name)
```

---

## range()

Generate a sequence of numbers.

```python
range(stop)
range(start, stop)
range(start, stop, step)
```

Examples:

```python
range(5)

range(2, 8)

range(1, 10, 2)
```

---

## enumerate()

Get both index and value.

```python
names = ["Alice", "Bob"]

for index, name in enumerate(names):
    print(index, name)
```

---

## zip()

Iterate over multiple iterables simultaneously.

```python
names = ["Alice", "Bob"]
scores = [90, 85]

for name, score in zip(names, scores):
    print(name, score)
```

---

# 3. while Loops

A `while` loop repeats while its condition remains `True`.

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

---

## Infinite Loops

```python
while True:
    ...
```

Terminate using:

```python
break
```

---

## Sentinel Loops

```python
while True:
    text = input("> ")

    if text == "quit":
        break

    print(text)
```

---

# 4. Loop Control Statements

## break

Exit the nearest loop immediately.

```python
for i in range(10):
    if i == 5:
        break

    print(i)
```

---

## continue

Skip the current iteration.

```python
for i in range(5):

    if i == 2:
        continue

    print(i)
```

Output:

```
0
1
3
4
```

---

## pass

Placeholder statement.

```python
def func():
    pass
```

---

# 5. Loop else Clause

## for...else

The `else` block runs only if the loop finishes without `break`.

```python
for item in items:

    if item == target:
        break

else:
    print("Not found")
```

---

## while...else

Works the same way.

```python
while x < 5:
    x += 1
else:
    print("Finished")
```

# 6. Comprehension Syntax

Comprehensions create collections using concise syntax.

---

## List Comprehension

```python
squares = [x * x for x in range(1, 6)]
```

---

### Filtering

```python
evens = [x for x in range(10) if x % 2 == 0]
```

---

### Conditional Expression

```python
values = [
    x if x > 0 else 0
    for x in nums
]
```

---

## Dictionary Comprehension

```python
squares = {
    x: x * x
    for x in range(5)
}
```

---

## Set Comprehension

```python
unique_lengths = {
    len(word)
    for word in words
}
```

---

## Generator Expression

```python
numbers = (
    x * x
    for x in range(10)
)
```

Unlike list comprehensions, generator expressions produce values lazily.

---

# 7. Common Pitfalls

- Using `=` instead of `==`
- Forgetting to update a `while` loop variable
- Creating accidental infinite loops
- Off-by-one errors with `range()`
- Modifying a list while iterating over it
- Confusing `break` and `continue`
- Forgetting that comprehensions always iterate over an iterable

---

# 8. Best Practices

- Prefer `for` when iterating over collections.
- Use `while` when the number of iterations is unknown.
- Keep conditions simple and readable.
- Use comprehensions for simple transformations.
- Avoid deeply nested conditionals.
- Use meaningful variable names.
- Ensure every `while` loop has a clear exit condition.

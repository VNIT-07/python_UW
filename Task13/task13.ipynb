# 1. Python Standard Library

## What is a Library?

A library is a collection of pre-written Python code that you can import and reuse.

Example:

```python
import math
print(math.sqrt(25))
```

## Standard Library vs Third-Party Library

| Feature | Standard Library | Third-Party Library |
|--------|------------------|---------------------|
| Installation | Comes with Python | Install using `pip` |
| Example | `math`, `os`, `random` | `numpy`, `pandas`, `torch` |

# 2. math Module

```python
import math
```

Used for mathematical calculations.

| Method | Syntax | Returns | Purpose | Example | Common Use |
|--------|--------|---------|---------|---------|------------|
| `pi` | `math.pi` | `float` | Pi constant | `math.pi` | Geometry |
| `e` | `math.e` | `float` | Euler's constant | `math.e` | Statistics |
| `sqrt()` | `math.sqrt(x)` | `float` | Square root | `math.sqrt(25)` → `5.0` | Distance formulas |
| `ceil()` | `math.ceil(x)` | `int` | Round upward | `math.ceil(4.2)` → `5` | Pagination |
| `floor()` | `math.floor(x)` | `int` | Round downward | `math.floor(4.9)` → `4` | Indexing |
| `fabs()` | `math.fabs(x)` | `float` | Absolute value | `math.fabs(-7)` → `7.0` | Numeric calculations |
| `factorial()` | `math.factorial(n)` | `int` | Factorial | `math.factorial(5)` → `120` | Probability |
| `gcd()` | `math.gcd(a,b)` | `int` | Greatest common divisor | `math.gcd(18,24)` → `6` | Number theory |
| `lcm()` | `math.lcm(a,b)` | `int` | Least common multiple | `math.lcm(12,18)` → `36` | Scheduling |
| `pow()` | `math.pow(a,b)` | `float` | Exponentiation | `math.pow(2,3)` → `8.0` | Math |
| `log()` | `math.log(x,base)` | `float` | Logarithm | `math.log(16,2)` → `4` | ML |
| `log10()` | `math.log10(x)` | `float` | Base-10 log | `math.log10(1000)` → `3` | Scientific computing |
| `sin()` | `math.sin(x)` | `float` | Sine | `math.sin(math.pi/2)` | Trigonometry |
| `cos()` | `math.cos(x)` | `float` | Cosine | `math.cos(0)` | Trigonometry |
| `tan()` | `math.tan(x)` | `float` | Tangent | `math.tan(math.pi/4)` | Trigonometry |
| `radians()` | `math.radians(x)` | `float` | Degrees → Radians | `math.radians(90)` | Angle conversion |
| `degrees()` | `math.degrees(x)` | `float` | Radians → Degrees | `math.degrees(math.pi/2)` | Angle conversion |
| `isclose()` | `math.isclose(a,b)` | `bool` | Compare floats safely | `math.isclose(0.1+0.2,0.3)` | Float comparison |

# 3. random Module

```python
import random
```

| Method | Syntax | Returns | Purpose | Example | Notes |
|--------|--------|---------|---------|---------|-------|
| `random()` | `random.random()` | `float` | Random float | `0.53` | Range `[0,1)` |
| `randint()` | `random.randint(a,b)` | `int` | Random integer | `randint(1,6)` | Inclusive |
| `randrange()` | `random.randrange(a,b)` | `int` | Random integer | `randrange(1,10)` | Stop excluded |
| `uniform()` | `random.uniform(a,b)` | `float` | Random decimal | `uniform(1,5)` | Float range |
| `choice()` | `random.choice(seq)` | Element | Pick one | `choice(colors)` | Single element |
| `choices()` | `random.choices(seq,k=n)` | `list` | Pick many | `choices(nums,k=5)` | Duplicates allowed |
| `sample()` | `random.sample(seq,k)` | `list` | Pick unique | `sample(nums,3)` | No duplicates |
| `shuffle()` | `random.shuffle(lst)` | `None` | Shuffle list | `shuffle(cards)` | In-place |
| `seed()` | `random.seed(42)` | `None` | Fixed randomness | `seed(42)` | Reproducible |

# 4. datetime Module

```python
from datetime import date, datetime, timedelta
```

| Method | Syntax | Returns | Purpose | Example |
|--------|--------|---------|---------|---------|
| `date.today()` | `date.today()` | `date` | Current date | `2026-07-23` |
| `datetime.now()` | `datetime.now()` | `datetime` | Current date & time | `datetime.now()` |
| `date()` | `date(y,m,d)` | `date` | Create date | `date(2026,7,23)` |
| `datetime()` | `datetime(y,m,d,h,m)` | `datetime` | Create datetime | `datetime(2026,7,23,10,30)` |
| `strftime()` | `dt.strftime(fmt)` | `str` | Format date | `"%d-%m-%Y"` |
| `strptime()` | `datetime.strptime(s,fmt)` | `datetime` | Parse string | `"22-07-2026"` |
| `timedelta()` | `timedelta(days=7)` | `timedelta` | Date arithmetic | `today+timedelta(days=7)` |
| `.days` | `delta.days` | `int` | Days difference | `(d2-d1).days` |

# 5. os Module

```python
import os
```

## Core Methods

| Method | Syntax | Returns | Purpose | Example |
|--------|--------|---------|---------|---------|
| `getcwd()` | `os.getcwd()` | `str` | Current directory | `os.getcwd()` |
| `chdir()` | `os.chdir(path)` | `None` | Change directory | `os.chdir("..")` |
| `listdir()` | `os.listdir()` | `list` | List files | `os.listdir()` |
| `mkdir()` | `os.mkdir(name)` | `None` | Create folder | `os.mkdir("logs")` |
| `makedirs()` | `os.makedirs(path, exist_ok=True)` | `None` | Create nested folders | `os.makedirs(...)` |
| `rename()` | `os.rename(old,new)` | `None` | Rename | `os.rename("a","b")` |
| `remove()` | `os.remove(file)` | `None` | Delete file | `os.remove("a.txt")` |
| `rmdir()` | `os.rmdir(folder)` | `None` | Delete empty folder | `os.rmdir("logs")` |

## os.path

| Method | Syntax | Returns | Purpose | Example |
|--------|--------|---------|---------|---------|
| `exists()` | `os.path.exists(path)` | `bool` | Check existence | `exists("a.txt")` |
| `isfile()` | `os.path.isfile(path)` | `bool` | Is file? | `isfile("a.txt")` |
| `isdir()` | `os.path.isdir(path)` | `bool` | Is directory? | `isdir("data")` |
| `join()` | `os.path.join(a,b)` | `str` | Join paths | `join("data","train.csv")` |
| `abspath()` | `os.path.abspath(path)` | `str` | Absolute path | `abspath("a.txt")` |
| `basename()` | `os.path.basename(path)` | `str` | File name | `basename(path)` |
| `dirname()` | `os.path.dirname(path)` | `str` | Parent folder | `dirname(path)` |
| `expanduser()` | `os.path.expanduser("~")` | `str` | Home directory | `expanduser("~")` |

### II. What is Python Interpreter

#### A **Python interpreter** is a program that reads, translates, and executes Python code line by line.

#### **=> How it works.**

- You write Python code in a .py /.ipynb file.

- The Python interpreter reads the code.

- It converts the code into bytecode.

- The Python Virtual Machine (PVM) executes the bytecode.

- The program produces the output.

```python

words = []

while True:
    wrd_inp = input("Enter any word: ")
    if wrd_inp == "0":
        break
    words.append(wrd_inp)
    res = [len(i) for i in words]
    print(res)
    print(words)
    
```

```python
words = []

while True:
    wrd_inp = input("Enter any word: ")
    if wrd_inp == "0":
        break
    words.append(wrd_inp)
    res = [i.title() for i in words]
    print(res)
```

```python
mat = [1,2,3],[4,5,6],[7,8,9]

res = [items for i in mat for items in i]
print(res)
```

```python
sentence = "python is fun and python is powerful"

words = sentence.split()

frequency = {
    word: words.count(word)
    for word in set(words)
}

print(frequency)
```

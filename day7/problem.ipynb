## Problem 1 – Squares

Create a list containing squares from **1 to 20**.

```python
[x*x for x in range(1,21)]
```

## Problem 2 – Even Numbers

Generate all even numbers from **1 to 50**.

```python
[x for x in range(1,51) if x % 2 == 0]
```

## Problem 3 – Filter Positive Numbers

```python
nums = [5,-3,8,-1,0,9,-10]

only_positive = [i for i in nums if i >= 0]
```

## Problem 4 – Replace Negative Numbers with Zero

```python
nums = [5,-3,8,-1,0,9,-10]

replaced = [i if i >= 0 else 0 for i in nums]
```

## Problem 5 – Length of Words

## Problem 7 – Flatten Matrix

```python
matrix = [
    [1,2,3],
    [4,5],
    [6,7,8,9]
]

flat = [item for row in matrix for item in row]
```

## Problem 8 – Cartesian Product

```python
colors = ["Red","Blue"]
sizes = ["S","M","L"]

pairs = [(color,size) for color in colors for size in sizes]
```

## Problem 9 – Multiplication Table

```python
n = 7

table = [
    f"{n} * {i} = {n*i}"
    for i in range(1,11)
]
```

## Problem 10 – Filter PNG Files

```python
files = [
    "cat.png",
    "dog.jpg",
    "report.pdf",
    "flower.png",
    "movie.mp4"
]

png_files = [
    file
    for file in files
    if file.endswith(".png")
]
```

## Problem 11 – Matrix Transpose

```python
matrix = [
    [1,2,3],
    [4,5,6]
]

transpose = [
    [row[i] for row in matrix]
    for i in range(len(matrix[0]))
]
```

## Problem 12 – Prime Numbers

```python
primes = [
    p
    for p in range(2,101)
    if all(
        p % d != 0
        for d in range(2,int(p**0.5)+1)
    )
]
```

## Problem 13 – Remove Duplicate Words

```python
words = [
    "AI",
    "ML",
    "AI",
    "Python",
    "ML"
]

unique = [
    words[i]
    for i in range(len(words))
    if all(
        words[w] != words[i]
        for w in range(i)
    )
]
```

## Problem 14 – Dictionary to Strings

```python
student = {
    "name":"Vatsal",
    "age":21,
    "cgpa":8.5
}

result = [
    f"{k} = {v}"
    for k,v in student.items()
]
```

## Problem 16 – Flatten Only Even Numbers

```python
matrix = [
    [1,2,3],
    [4,5],
    [6,7,8,9]
]

flat_even = [
    item
    for row in matrix
    for item in row
    if item % 2 == 0
]
```

## Problem 17 – FizzBuzz

```python
result = [
    "FizzBuzz" if n % 15 == 0
    else "Fizz" if n % 3 == 0
    else "Buzz" if n % 5 == 0
    else n
    for n in range(1,51)
]
```

## Problem 18 – Word Frequency Dictionary

```python
sentence = "python is fun and python is powerful"

words = sentence.split()

frequency = {
    word: words.count(word)
    for word in set(words)
}
```

## Problem 19 – Conditional Cartesian Product

```python
[
    (i,j)
    for i in range(1,6)
    for j in range(1,6)
    if i != j and (i+j)%2==0
]
```

## Problem 20 – Spiral Pattern

```python
spiral = [
    [j for j in range(1,i+1)]
    for i in range(1,6)
]
```

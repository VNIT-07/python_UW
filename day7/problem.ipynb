## Problem 11 – Matrix Transpose

## Problem 12 – Prime Numbers

```python
primes = [
    p
    for p in range(2,101)
    if all(
        p % d != 0
        for d in range(2,int(p**0.5)+1)
    )
]
```

## Problem 13 – Remove Duplicate Words

```python
words = [
    "AI",
    "ML",
    "AI",
    "Python",
    "ML"
]

unique = [
    words[i]
    for i in range(len(words))
    if all(
        words[w] != words[i]
        for w in range(i)
    )
]
```

```python
result = [
    "FizzBuzz" if n % 15 == 0
    else "Fizz" if n % 3 == 0
    else "Buzz" if n % 5 == 0
    else n
    for n in range(1,51)
]
```

```python
sentence = "python is fun and python is powerful"

words = sentence.split()

frequency = {
    word: words.count(word)
    for word in set(words)
}
```

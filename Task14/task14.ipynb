```python
class AgeError(Exception):
    pass


try: 
    age  = int(input("enter your age! "))
    
    if(age < 18):
        raise AgeError
    print("Db connection done!")

except ValueError:
    
    print("invalide age")
except AgeError:
    print("age should be > 18 !")
except Exception:
    print("something went wrong")
finally:
    print("Db connection close!")
```

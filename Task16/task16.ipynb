## 1. Iterables

An **iterable** is any object that can be looped over. Examples: list, tuple, string, dictionary, set, range.

Internally, an iterable implements `__iter__()` which returns an iterator.

````python
numbers = [10, 20, 30]

for n in numbers:
    print(n)

print(hasattr(numbers, "__iter__"))
print(hasattr(numbers, "__next__"))
````

### Common Iterables

````python
items = [
    [1,2,3],
    (1,2,3),
    "Python",
    {"a":1, "b":2},
    {1,2,3},
    range(5)
]

for obj in items:
    print(type(obj).__name__, "is iterable?", hasattr(obj, "__iter__"))
````

## 2. Iterators

An **iterator** remembers its current position and produces one value at a time using `next()`.

Iterator methods:
- `__iter__()`
- `__next__()`

````python
numbers = [10,20,30]

it = iter(numbers)

print(next(it))
print(next(it))
print(next(it))

try:
    print(next(it))
except StopIteration:
    print("Iteration finished!")
````

### Iterator State

````python
text = "ABC"
it = iter(text)

print(next(it))
print(next(it))
print(next(it))
````

## 3. Iterables vs Iterators

| Iterable | Iterator |
|---|---|
|Can be looped over|Produces values one by one|
|Uses `iter()` to create iterator|Uses `next()`|
|Doesn't remember position|Remembers current position|
|Examples: list, tuple, string|Created using `iter()` or generators|

````python
nums = {1,2,3}
iterator = iter(nums)

print(type(nums))
print(type(iterator))
````

## 4. Generators

A generator is a special iterator created using the `yield` keyword. It generates values lazily (on demand), making it memory efficient.

````python
def countdown(n):
    while n > 0:
        yield n
        n -= 1

gen = countdown(5)

# for value in gen:
#     print(value)
print(next(gen))
print(next(gen))
````

### yield vs return

````python
def demo_return():
    return 10, 10

def demo_yield():
    yield 10, 10
    # yield 20

print(demo_return())
print(list(demo_yield()))
````

### Memory Efficiency

````python
def square_generator(n):
    for i in range(n):
        yield i*i

for value in square_generator(5):
    print(value)
````

````python
squares = [x*x for x in range(6)]

# print(next(squares))
# print(next(squares))

# for x in squares:
#     print(x)
iterable = iter(squares)
print(next(iterable))
````

## 6. Decorators

A decorator modifies or extends the behavior of another function without changing its source code.

Syntax:
```python
@decorator_name
def func():
    ...
```

````python
def decorator(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")
    return wrapper

@decorator
def greet():
    print("Hello!")

greet()
````

### Decorator with Arguments

````python
def repeat(func):
    def wrapper(*args, **kwargs):
        func(*args, **kwargs)
        func(*args, **kwargs)
    return wrapper

@repeat
def say(name):
    print("Hello", name)

say("Alice")
````

### Using functools.wraps

````python
from functools import wraps

def logger(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@logger
def add(a,b):
    return a+b

print(add(2,3))
print(add.__name__)
````

## 7. Practical Applications

### Logging Decorator

````python
from functools import wraps

def log(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"[LOG] Executing {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@log
def multiply(a,b):
    return a*b

print(multiply(4,5))
````

### Timing Decorator

````python
import time
from functools import wraps

def timer(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        print("Elapsed:", round(time.time()-start,6),"seconds")
        return result
    return wrapper

@timer
def work():
    total = sum(range(100000))
    return total

print(work())
````

### Reading Large Files with Generator

````python
def read_lines(lines):
    for line in lines:
        yield line.strip()

sample = ["Python\n","Java\n","C++\n"]

for line in read_lines(sample):
    print(line)
````

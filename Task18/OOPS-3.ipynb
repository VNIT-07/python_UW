# 1. What is Encapsulation?

Encapsulation is one of the four pillars of Object-Oriented Programming.

A textbook definition is:

> Bundling data and methods together while restricting direct access to some of the object's internal state.

A more practical definition is:

> **Encapsulation is controlling how an object's data is accessed and modified.**

Instead of allowing everyone to directly modify an object's data, we expose controlled operations.

---

# Real-Life Example

Think about an ATM.

```
Customer
     │
     ▼
 Deposit ₹500
     │
     ▼
Bank Account
```

You cannot directly modify the balance stored inside the bank's database.

Instead, you use controlled operations:

- Deposit
- Withdraw
- Check Balance

The bank decides whether your request is valid.

Objects work the same way.

---

# 2. Why Encapsulation?

Without encapsulation:

```python
account.balance = -100000
```

Now the object contains invalid data.

With encapsulation:

```python
account.withdraw(1000)
```

The object validates the request before changing its internal state.

Benefits:

- Protect object state
- Prevent invalid data
- Hide implementation details
- Improve maintainability
- Make debugging easier

# 3. Access Levels in Python

Unlike Java or C++, Python does not have true access modifiers.

Instead, Python follows naming conventions.

There are three commonly used levels.

| Type | Prefix | Accessibility |
|-------|--------|---------------|
| Public | `name` | Accessible everywhere |
| Protected | `_name` | Accessible, but intended for internal use |
| Private | `__name` | Name-mangled to reduce accidental access |

---

# 4. Public Members

Public members are accessible from anywhere.

Example

```python
class Student:

    def __init__(self):
        self.name = "Alice"

s = Student()

print(s.name)
```

Output

```
Alice
```

Public members are the default in Python.

---

# 5. Protected Members

Protected members use a single underscore.

```python
class Student:

    def __init__(self):
        self._marks = 95
```

Accessing it

```python
print(student._marks)
```

works perfectly.

---

## Does Protected Prevent Access?

No.

Protected members **do not** prevent access.

The underscore simply tells other developers:

> "This is an internal implementation detail. Please don't access it unless necessary."

It is a convention, not a security feature.

---

## When to Use Protected Members

Protected members are useful when:

- subclasses need access
- external code generally should not modify the value
- you want to communicate intent

Example

```python
class Animal:

    def __init__(self):
        self._energy = 100
```

Subclasses can safely use `_energy`.

---

# 6. Private Members

Private members begin with two underscores.

```python
class Student:

    def __init__(self):
        self.__name = "Alice"
```

Trying

```python
print(student.__name)
```

produces

```
AttributeError
```

At first glance, this looks like Python has made the variable private.

However, something different is happening internally.

---

# 7. Name Mangling

Python performs **name mangling** for attributes beginning with two underscores.

Internally,

```python
self.__name
```

becomes approximately

```python
self._Student__name
```

where `Student` is the class name.

This transformation reduces accidental access and prevents name collisions in subclasses.

---

## Example

```python
class Student:

    def __init__(self):
        self.__name = "Alice"

student = Student()

print(student.__dict__)
```

Output

```
{
    '_Student__name': 'Alice'
}
```

The attribute still exists.

Its name has changed.

---

# 8. Accessing a Mangled Attribute

Although not recommended,

you can access the attribute using its mangled name.

```python
print(student._Student__name)
```

Output

```
Alice
```

This demonstrates that Python's private members are not truly private.

They are intentionally harder to access.

---

# 9. Name Mangling and Inheritance

Consider

```python
class Person:

    def __init__(self):
        self.__name = "Person"
```

```python
class Student(Person):

    def __init__(self):

        super().__init__()

        self.__name = "Student"
```

Inspecting

```python
print(student.__dict__)
```

produces

```
{
    '_Person__name': 'Person',
    '_Student__name': 'Student'
}
```

Notice that both variables coexist.

This is one of the primary reasons Python uses name mangling.

It prevents subclasses from accidentally overwriting parent attributes.

# 10. Public vs Protected vs Private

| Feature | Public | Protected | Private |
|---------|---------|-----------|----------|
| Prefix | `name` | `_name` | `__name` |
| Accessible Everywhere | Yes | Yes | No (directly) |
| Name Mangling | No | No | Yes |
| Prevents Accidental Collision | No | No | Yes |
| Intended Use | General | Internal | Strong internal implementation |

# 11. Getters

A getter provides controlled read access.

Example

```python
class BankAccount:

    def __init__(self):
        self.__balance = 1000

    def get_balance(self):
        return self.__balance
```

Usage

```python
account.get_balance()
```

Instead of exposing the variable directly, we expose a controlled method.

---

# 12. Setters

A setter validates data before updating it.

Example

```python
class BankAccount:

    def __init__(self):
        self.__balance = 1000

    def set_balance(self, amount):

        if amount >= 0:
            self.__balance = amount
```

Now invalid values can be rejected.

---

# 13. Traditional Getters and Setters

Many languages use

```python
obj.get_balance()

obj.set_balance(500)
```

Python supports this approach,

but it offers a more elegant alternative.

---

# 14. The @property Decorator

`@property` allows a method to behave like an attribute.

Example

```python
class Temperature:

    def __init__(self):
        self.__celsius = 25

    @property
    def celsius(self):
        return self.__celsius
```

Usage

```python
temp = Temperature()

print(temp.celsius)
```

Notice

```
temp.celsius
```

looks like a variable,

but Python internally calls the method.

---

# 15. Property Setter

A setter is attached using

```python
@property_name.setter
```

Example

```python
class Temperature:

    def __init__(self):
        self.__celsius = 25

    @property
    def celsius(self):
        return self.__celsius

    @celsius.setter
    def celsius(self, value):

        if value >= -273.15:
            self.__celsius = value
```

Usage

```python
temp.celsius = 30
```

Internally,

Python executes

```python
temp.celsius(30)
```

behind the scenes.

The syntax remains clean while validation stays centralized.

---

# 16. Read-Only Properties

If only the getter exists,

```python
@property
def balance(self):
    return self.__balance
```

then

```python
account.balance = 500
```

raises

```
AttributeError
```

This creates a read-only property.

---

# 17. Computed Properties

Properties do not have to return stored data.

They can compute values dynamically.

Example

```python
class Rectangle:

    def __init__(self, width, height):
        self.width = width
        self.height = height

    @property
    def area(self):
        return self.width * self.height
```

Usage

```python
r = Rectangle(4, 5)

print(r.area)
```

Output

```
20
```

No `area` variable is stored.

The value is calculated each time it is requested.

---

# 18. Common Beginner Mistakes

### Mistake 1

Assuming private means secure.

Private members only undergo name mangling.

---

### Mistake 2

Using double underscores everywhere.

Most internal attributes should simply be protected.

Reserve private members for avoiding accidental collisions or hiding implementation details.

---

### Mistake 3

Using setters without validation.

A setter should exist to enforce rules.

Otherwise, direct assignment may be simpler.

---

### Mistake 4

Ignoring `@property`.

Pythonic code generally prefers

```python
obj.balance
```

over

```python
obj.get_balance()
```

when appropriate.

---

# 19. Best Practices

- Default to public attributes when appropriate.
- Use protected members for internal implementation details.
- Use private members only when necessary.
- Prefer `@property` over explicit getters and setters.
- Keep validation logic inside setters.
- Avoid exposing internal implementation details unnecessarily.

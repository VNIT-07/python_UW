# 1. What is Object-Oriented Programming (OOP)?

Object-Oriented Programming (OOP) is a programming paradigm in which a program is organized around **objects** rather than individual functions.

Instead of asking:

> "What functions should I write?"

OOP asks:

> "What objects exist in my system and what should they know and do?"

Objects combine:

- Data (Attributes)
- Behavior (Methods)

into one logical unit.

---

## Real-Life Analogy

Think about a **Car**.

A car has:

### Data (Attributes)

- Brand
- Model
- Color
- Speed
- Fuel

and it can perform actions:

### Behaviors (Methods)

- Start
- Stop
- Accelerate
- Brake
- Turn

A real car naturally groups its data and behavior together.

OOP follows the same philosophy.

---

# 2. Why Do We Need OOP?

Suppose we want to manage information about students.

Without OOP:

```python
student1_name = "Alice"
student1_age = 20
student1_marks = 92

student2_name = "Bob"
student2_age = 21
student2_marks = 88
```

Imagine storing information for **10,000 students**.

The code quickly becomes difficult to maintain.

Instead, OOP allows us to define one blueprint.

```python
class Student:
    pass
```

Then create as many students as needed.

```python
s1 = Student()
s2 = Student()
s3 = Student()
```

This makes programs:

- Reusable
- Organized
- Easier to maintain
- Easier to extend

# 3. Procedural Programming vs OOP

## Procedural Programming

Focus:

> Functions operate on data.

Example:

```python
name = "Alice"
marks = 95

def display(name, marks):
    print(name, marks)

display(name, marks)
```

Data and functions are separate.

---

## Object-Oriented Programming

Focus:

> Objects own their own data and behavior.

Example:

```python
class Student:

    def display(self):
        print(self.name, self.marks)
```

The object manages itself.

---

## Comparison

| Procedural Programming | Object-Oriented Programming |
|------------------------|-----------------------------|
| Data and functions are separate | Data and functions stay together |
| Difficult to scale | Easier to scale |
| More repetition | Better code reuse |
| Less modular | Highly modular |
| Suitable for small scripts | Suitable for medium to large applications |

# 4. The Four Pillars of OOP

Python's Object-Oriented Programming is built on four major concepts.

## 1. Encapsulation

Bundling data and methods together while controlling access to internal data.

Example:

```python
class BankAccount:
    def deposit(self):
        ...
```

---

## 2. Inheritance

Creating a new class from an existing class.

Example:

```python
class Animal:
    pass

class Dog(Animal):
    pass
```

The child inherits the parent's features.

---

## 3. Polymorphism

One interface, many implementations.

Example:

```python
animal.speak()
```

Depending on the object:

```
Dog → Bark

Cat → Meow

Cow → Moo
```

---

## 4. Abstraction

Showing only essential features while hiding implementation details.

Python supports abstraction primarily using abstract base classes (`abc` module), which you'll encounter later in advanced OOP.

# 5. Class

A **class** is a blueprint used to create objects.

Think of a class as an architectural drawing.

The blueprint itself is not the house.

It only describes how houses should be built.

Similarly,

A class only describes what future objects will contain.

---

## Example

```python
class Student:
    pass
```

This creates a blueprint.

No student exists yet.

# 6. Object

An object is an actual instance created from a class.

Example:

```python
class Student:
    pass

s1 = Student()
s2 = Student()
```

Now we have two different objects.

Even though both were created from the same blueprint, they occupy different memory locations.

---

## Memory Representation

```
Student Class
      │
      ▼
 ┌────────────┐
 │ Blueprint  │
 └────────────┘
      │
      ▼
 ┌──────────┐
 │    s1    │
 └──────────┘

 ┌──────────┐
 │    s2    │
 └──────────┘
```

---

# 7. Class vs Object

| Class | Object |
|--------|--------|
| Blueprint | Real instance |
| Defined once | Can create many |
| Doesn't store real data | Stores actual values |
| Logical structure | Physical entity in memory |

# 9. Understanding the self Keyword

The `self` parameter is one of the most important concepts in Python OOP.

Many beginners think it is a keyword.

It is **not**.

It is simply a naming convention.

You could write:

```python
class Student:

    def greet(me):
        print("Hello")
```

Python accepts this.

However, using `self` is the universal convention.

---

## Why Do We Need self?

Suppose we create two students.

```python
s1 = Student()
s2 = Student()
```

Each object has its own data.

```
s1
name = Alice

s2
name = Bob
```

When we execute:

```python
s1.display()
```

Python must know which object's data should be used.

It automatically passes the object as the first argument.

Internally:

```python
Student.display(s1)
```

Likewise,

```python
s2.display()
```

becomes

```python
Student.display(s2)
```

Therefore,

inside the method,

```python
self
```

refers to the object that called the method.

# 10. Instance Variables

Instance variables belong to individual objects.

They are created using:

```python
self.variable = value
```

Example:

```python
class Student:

    def __init__(self, name):
        self.name = name
```

Creating objects:

```python
s1 = Student("Alice")
s2 = Student("Bob")
```

Memory:

```
s1
└── name = Alice

s2
└── name = Bob
```

Each object owns its own copy.

Changing one object does not affect another.

---

# 11. Constructors (__init__)

A constructor initializes an object immediately after it is created.

In Python, constructors are written using:

```python
__init__()
```

Example:

```python
class Student:

    def __init__(self):
        print("Object Created")
```

Creating an object:

```python
s = Student()
```

Output:

```
Object Created
```

---

# 12. Parameterized Constructors

Constructors often receive information during object creation.

Example:

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age
```

Creating objects:

```python
s1 = Student("Alice", 20)
s2 = Student("Bob", 22)
```

Now every object stores different values.

---

# 13. Default Parameters

Constructors can also use default arguments.

```python
class Student:

    def __init__(self, name, age=18):
        self.name = name
        self.age = age
```

Usage:

```python
Student("Alice")
Student("Bob", 21)
```

---

# 14. Object Lifecycle

Every object goes through a simple lifecycle.

```
Class
   │
   ▼
Object Created
   │
   ▼
Constructor Runs
   │
   ▼
Object Used
   │
   ▼
Object Destroyed (Garbage Collection)
```

Python automatically frees memory when objects are no longer referenced.

# 15. Common Beginner Mistakes

### Forgetting `self`

Incorrect:

```python
class Student:

    def greet():
        print("Hello")
```

Correct:

```python
class Student:

    def greet(self):
        print("Hello")
```

---

### Forgetting `self.`

Incorrect:

```python
class Student:

    def __init__(self, name):
        name = name
```

This creates a local variable.

Correct:

```python
self.name = name
```

---

### Calling `__init__()` Manually

Incorrect:

```python
s = Student()

s.__init__()
```

The constructor is called automatically when the object is created.

---

# 16. Best Practices

- Use meaningful class names (`Student`, `BankAccount`, `Employee`).
- Use PascalCase for class names.
- Use `self` consistently.
- Initialize instance variables inside `__init__()`.
- Keep constructors focused on initialization only.
- Avoid writing heavy business logic inside constructors.

# 1. What is Polymorphism?

The word **Polymorphism** comes from two Greek words:

- **Poly** = Many
- **Morph** = Forms

Therefore,

> **Polymorphism means "One Interface, Many Forms."**

Different objects can respond differently to the same method call.

---

# Real-Life Example

Imagine a remote control.

```
          Power Button
                │
     ┌──────────┼──────────┐
     │          │          │
     ▼          ▼          ▼
    TV         AC      Speaker
```

You press the same button.

Each device performs a different action.

- TV → Turns On
- AC → Starts Cooling
- Speaker → Plays Music

The interface remains the same.

The behavior changes.

This is polymorphism.

# 2. Runtime Polymorphism

Python determines which method should execute while the program is running.

Example

```python
class Animal:

    def speak(self):
        print("Animal Sound")


class Dog(Animal):

    def speak(self):
        print("Bark")


class Cat(Animal):

    def speak(self):
        print("Meow")
```

Usage

```python
animals = [Dog(), Cat(), Animal()]

for animal in animals:
    animal.speak()
```

Output

```
Bark
Meow
Animal Sound
```

The loop never checks the object's type.

Python automatically executes the correct method.

---

# 3. Method Overriding

Method overriding occurs when a child class provides its own implementation of an inherited method.

Parent

```python
class Animal:

    def speak(self):
        print("Animal")
```

Child

```python
class Dog(Animal):

    def speak(self):
        print("Bark")
```

The child's version replaces the parent's implementation.

---

# 4. Extending Parent Behavior

Sometimes we want to reuse the parent's behavior before adding new functionality.

```python
class Animal:

    def speak(self):
        print("Animal Sound")


class Dog(Animal):

    def speak(self):
        super().speak()
        print("Bark")
```

Output

```
Animal Sound
Bark
```

# 5. Duck Typing

Duck Typing is a Python feature where an object's type is less important than its behavior.

Python follows the philosophy:

> "If it walks like a duck and quacks like a duck, treat it as a duck."

---

## Example

```python
class Dog:

    def speak(self):
        print("Bark")


class Robot:

    def speak(self):
        print("Beep")


def make_sound(obj):
    obj.speak()


make_sound(Dog())
make_sound(Robot())
```

Output

```
Bark
Beep
```

Neither class inherits from the other.

Python only checks whether the object has the required method.

---

# 6. Why Duck Typing Matters

Many Python libraries use Duck Typing extensively.

Example

```python
def save(file):
    file.write("Hello")
```

Python doesn't care whether `file` is:

- a text file
- a PDF writer
- an in-memory buffer
- a network stream

It only requires the object to implement

```python
write()
```

# 7. Real AI/ML Example

Machine learning libraries follow a common interface.

```python
LinearRegression.predict()

DecisionTree.predict()

RandomForest.predict()
```

A function can work with any model.

```python
def evaluate(model, X):
    return model.predict(X)
```

The function never checks the model's type.

It simply calls `predict()`.

This is polymorphism.

# 9. Common OOP Mistakes

## Mistake 1

Using inheritance for unrelated classes.

Incorrect

```
Car → Person
```

Correct

```
Car → Vehicle

Dog → Animal
```

Inheritance should represent an **"is-a"** relationship.

---

## Mistake 2

Using private members everywhere.

Most internal variables should simply be protected.

Use private members only when avoiding accidental access or name collisions.

---

## Mistake 3

Ignoring constructors.

Always initialize important instance variables inside `__init__()`.

---

## Mistake 4

Writing large constructors.

Constructors should initialize an object.

Avoid complex business logic inside them.

---

## Mistake 5

Creating God Classes.

Avoid classes that perform dozens of unrelated tasks.

One class should have one primary responsibility.

---

## Mistake 6

Exposing every variable publicly.

Some object state should be protected through methods or properties.

---

## Mistake 7

Deep inheritance hierarchies.

Avoid structures like

```
A
│
B
│
C
│
D
│
E
│
F
```

Favor shallow hierarchies that are easier to understand.

# 14. Key Takeaways

- OOP organizes programs around objects.
- Classes define structure, while objects hold actual data.
- `self` always refers to the current object.
- Constructors initialize object state.
- Inheritance promotes code reuse.
- `super()` provides a safe way to extend parent behavior.
- Encapsulation protects an object's internal state.
- Python uses conventions (`_`) and name mangling (`__`) instead of strict access modifiers.
- `@property` enables clean, Pythonic interfaces with validation.
- Polymorphism allows different objects to respond to the same method call in different ways.
- Duck Typing emphasizes behavior over inheritance.

Mastering these concepts provides the foundation for advanced Python topics such as class methods, static methods, variable scopes, magic (dunder) methods, operator overloading, abstract base classes, design patterns, and large-scale application architecture.

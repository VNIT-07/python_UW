# 1. Why OOP?

### The Problem

Imagine you are building a school management system. Every student has a name, age, roll number, marks, and many operations such as studying, taking exams, and viewing results.

Without OOP you would create many unrelated variables, which quickly becomes difficult to maintain.

### The Solution

Object-Oriented Programming groups **data** and **behavior** into one unit called an **object**.

### Benefits
- Better organization
- Code reuse
- Easier maintenance
- Scalable applications
- Real-world modeling


````python
student1_name="Rahul"
student1_age=20
student2_name="Aman"
student2_age=22

print(student1_name, student2_name)
````

# 2. Class and Object

A **class** is a blueprint.

An **object** is a real instance created from that blueprint.

### Real-world example

Blueprint → House Design

Objects → House 1, House 2, House 3

One class can create thousands of objects.


````python
class Student:
    pass

s1=Student()
s2=Student()

print(type(s1))
print(s1 is s2)
````

# 3. Attributes

Attributes are variables stored inside an object.

Every object has its own copy of instance attributes.


````python
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age

s1=Student("Rahul",20)
s2=Student("Aman",22)

print(s1.__dict__)
print(s2.__dict__)
````

# 4. Methods

Methods are functions defined inside a class.

Methods describe what an object can do.


````python
class Student:
    def study(self):
        print("Studying Python")

Student().study()
````

# 5. Understanding self

`self` refers to the current object.

When you call:

```python
s1.study()
```

Python internally executes:

```python
Student.study(s1)
```

Therefore `self` becomes `s1`.


````python
class Demo:
    def show(self):
        print("Current object id:",id(self))

d=Demo()
d.show()
````

# 6. Constructor (__init__)

`__init__()` runs automatically whenever a new object is created.

Its purpose is to initialize object data.


````python
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age

print(Student("Rahul",20).__dict__)
````

# 7. Instance vs Class Variables

### Instance Variable
Different for every object.

### Class Variable
Shared by all objects.


````python
class Student:
    school="ABC School"

    def __init__(self,name):
        self.name=name

a=Student("Rahul")
b=Student("Aman")

print(a.school,b.school)
print(a.__dict__)
````

# 8. Encapsulation

Encapsulation combines data and methods and restricts direct access when appropriate.

Python conventions:
- Public → variable
- Protected → _variable
- Private → __variable

Private attributes should normally be accessed through methods or properties.


````python
class BankAccount:
    def __init__(self,balance):
        self.__balance=balance

    def deposit(self,amount):
        self.__balance+=amount

    def get_balance(self):
        return self.__balance

acc=BankAccount(1000)
acc.deposit(500)
print(acc.get_balance())
````

# 9. Abstraction

Abstraction hides unnecessary implementation details.

Users interact with a simple interface without knowing the internal logic.


````python
from abc import ABC,abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self,r):
        self.r=r
    def area(self):
        return 3.14*self.r*self.r

print(Circle(5).area())
````

# 10. Inheritance

Inheritance allows one class to reuse another class.

Relationship:
Dog **is an** Animal.


````python
class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    def bark(self):
        print("Barking")

d=Dog()
d.eat()
d.bark()
````

# 11. Polymorphism

Polymorphism means one interface, many implementations.

Different objects respond differently to the same method name.


````python
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

for animal in [Dog(),Cat()]:
    animal.sound()
````

# 12. Method Overriding & super()

A child class replaces a parent method.

Use `super()` when you also want the parent implementation.


````python
class Animal:
    def sound(self):
        print("Animal sound")

class Dog(Animal):
    def sound(self):
        super().sound()
        print("Dog Bark")

Dog().sound()
````

# 13. Multiple Inheritance & MRO

Python supports multiple inheritance.

Method Resolution Order (MRO) determines the search order.


````python
class A: pass
class B(A): pass
class C(A): pass
class D(B,C): pass

print(D.mro())
````

# 14. Composition

Composition represents a **has-a** relationship.

A Car has an Engine.


````python
class Engine:
    def start(self):
        print("Engine Started")

class Car:
    def __init__(self):
        self.engine=Engine()

Car().engine.start()
````

# 15. Class Methods, Static Methods, Properties

- `@classmethod` works with the class.
- `@staticmethod` is a utility function.
- `@property` provides controlled attribute access.


````python
class Student:
    school="ABC"

    @classmethod
    def change_school(cls,name):
        cls.school=name

    @staticmethod
    def add(a,b):
        return a+b

Student.change_school("XYZ")
print(Student.school)
print(Student.add(5,7))
````

# 16. Magic Methods

Magic (dunder) methods customize object behavior.

Examples:
- __str__
- __repr__
- __len__
- __eq__


````python
class Book:
    def __init__(self,title):
        self.title=title
    def __str__(self):
        return self.title

print(Book("Python"))
````

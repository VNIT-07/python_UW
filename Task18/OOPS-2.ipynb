# 1. What is Inheritance?

Inheritance allows one class to acquire the properties and behaviors of another class.

Instead of rewriting existing code, we reuse it.

Think of inheritance as creating a specialized version of an existing class.

---

## Real-Life Example

Consider a general class:

```
Animal
```

Every animal can:

- Eat
- Sleep
- Breathe

Now consider:

```
Dog
```

A dog can also:

- Bark
- Wag Tail

Should we rewrite:

- eat()
- sleep()
- breathe()

No.

Dog should simply inherit them.

---

## Diagram

```
           Animal
         /    |    \
       Dog   Cat   Cow
```

Every child receives common features from Animal.

---

# 2. Parent Class

The class whose properties are inherited is called the Parent Class.

Other common names:

- Base Class
- Superclass

Example

```python
class Animal:

    def eat(self):
        print("Eating")

    def sleep(self):
        print("Sleeping")
```

---

# 3. Child Class

The class that inherits another class is called the Child Class.

Other names:

- Derived Class
- Subclass

Example

```python
class Dog(Animal):

    def bark(self):
        print("Bark")
```

Notice

```python
class Dog(Animal)
```

The name inside parentheses is the parent class.

---

# 4. Using Inheritance

```python
class Animal:

    def eat(self):
        print("Eating")

class Dog(Animal):

    def bark(self):
        print("Bark")
```

Create object

```python
dog = Dog()
```

Calling inherited method

```python
dog.eat()
```

Output

```
Eating
```

Calling child method

```python
dog.bark()
```

Output

```
Bark
```

The Dog class contains only one method, yet it can call both methods because it inherits from Animal.

---

# 5. Memory Representation

```
Animal

eat()

sleep()

        ▲
        │
inherits
        │

Dog

bark()
```

Dog does not copy the code physically.

It simply gains access to the parent's methods.

---

# 6. Constructor Inheritance

Constructors are inherited only if the child does not define its own constructor.

Example

```python
class Animal:

    def __init__(self):
        print("Animal Created")
```

```python
class Dog(Animal):
    pass
```

Creating object

```python
dog = Dog()
```

Output

```
Animal Created
```

Dog automatically uses Animal's constructor.

---

# 7. Child Constructor

Suppose

```python
class Animal:

    def __init__(self):
        print("Animal")
```

```python
class Dog(Animal):

    def __init__(self):
        print("Dog")
```

Creating object

```python
Dog()
```

Output

```
Dog
```

The parent's constructor is **not** executed automatically.

The child constructor replaces it.

---

# 8. The super() Function

Sometimes we want both constructors.

Python provides

```python
super()
```

Example

```python
class Animal:

    def __init__(self):
        print("Animal Constructor")
```

```python
class Dog(Animal):

    def __init__(self):
        super().__init__()
        print("Dog Constructor")
```

Output

```
Animal Constructor
Dog Constructor
```

---

# 9. What Does super() Actually Do?

Many beginners think

```python
super()
```

means

> "Call my parent."

A more accurate explanation is:

> Return a proxy object that allows access to the next class in Python's Method Resolution Order (MRO).

For single inheritance, this effectively means the parent class.

---

## Internally

When you write

```python
super().__init__()
```

Python behaves approximately like

```python
Animal.__init__(self)
```

However,

`super()` is preferred because it also works correctly with multiple inheritance.

---

# 10. Passing Arguments to Parent Constructor

Example

```python
class Person:

    def __init__(self, name):
        self.name = name
```

```python
class Student(Person):

    def __init__(self, name, roll):

        super().__init__(name)

        self.roll = roll
```

Creating object

```python
s = Student("Alice", 101)
```

Memory

```
Student Object

name = Alice

roll = 101
```

The parent constructor initializes the inherited data.

The child constructor initializes its own data.

---

# 11. Constructor Chaining

Constructor execution follows a chain.

```
Student()

      │

      ▼

Student.__init__()

      │

super()

      │

      ▼

Person.__init__()
```

This is called **Constructor Chaining**.

It is one of the most common patterns in Python OOP.

---

# 12. Method Overriding

Sometimes the child wants different behavior.

Instead of using the inherited method, it replaces it.

Example

```python
class Animal:

    def speak(self):
        print("Animal Sound")
```

```python
class Dog(Animal):

    def speak(self):
        print("Bark")
```

Calling

```python
Dog().speak()
```

Output

```
Bark
```

The parent's method is overridden.

---

# 13. Calling Parent Method from Child

Sometimes we want both behaviors.

Example

```python
class Animal:

    def speak(self):
        print("Animal Sound")
```

```python
class Dog(Animal):

    def speak(self):

        super().speak()

        print("Bark")
```

Output

```
Animal Sound

Bark
```

This technique is useful when extending existing functionality.

---

# 14. Multiple Inheritance

Python supports inheriting from more than one parent.

Example

```python
class Camera:

    def click(self):
        print("Photo")
```

```python
class Phone:

    def call(self):
        print("Calling")
```

```python
class SmartPhone(Camera, Phone):
    pass
```

Usage

```python
s = SmartPhone()

s.call()

s.click()
```

Output

```
Calling

Photo
```

The child receives features from both parents.

---

# 15. Diamond Problem

Suppose

```
        A
      /   \
     B     C
      \   /
        D
```

If both

```
B
```

and

```
C
```

contain the same method,

which one should Python execute?

This ambiguity is called the Diamond Problem.

Python solves this using the Method Resolution Order.

---

# 16. Method Resolution Order (MRO)

Python searches for methods in a predefined order.

Example

```python
class A:
    pass

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass
```

Checking MRO

```python
print(D.mro())
```

Output

```
[D, B, C, A, object]
```

Whenever Python searches for a method, it follows this order.

---

# 17. Why Use super() Instead of ParentClass.method()?

Avoid

```python
Animal.__init__(self)
```

Prefer

```python
super().__init__()
```

Reasons

- Works with multiple inheritance.
- Respects MRO.
- Easier to maintain.
- More flexible for subclasses.

This is considered the Pythonic approach.

---

# 18. Common Beginner Mistakes

### Mistake 1

Forgetting to call

```python
super().__init__()
```

Result:

Parent variables are never initialized.

---

### Mistake 2

Using inheritance only to reuse a few lines of code.

Inheritance should model an **"is-a"** relationship.

Good

```
Dog is an Animal.
```

Bad

```
Car is a Person.
```

---

### Mistake 3

Overriding a method without understanding the parent's behavior.

Sometimes the parent performs essential work.

Use

```python
super().method()
```

when you need to extend rather than replace behavior.

---

# 19. Best Practices

- Prefer composition over inheritance when classes are not naturally related.
- Use inheritance only for true "is-a" relationships.
- Always call `super()` when extending constructors.
- Override methods only when behavior genuinely changes.
- Keep inheritance hierarchies shallow.

# What are Dunder Methods?

**Dunder** means **Double UNDERscore**.

Examples:

``` python
__init__
__str__
__repr__
__len__
__add__
__eq__
__getitem__
__call__
__iter__
```

They are called **special methods** because Python invokes them
automatically to implement built-in syntax.

------------------------------------------------------------------------

# How Python Uses Dunder Methods

  Python Syntax    Internal Call
  ---------------- ----------------------
  `print(obj)`     `obj.__str__()`
  `repr(obj)`      `obj.__repr__()`
  `len(obj)`       `obj.__len__()`
  `a + b`          `a.__add__(b)`
  `a == b`         `a.__eq__(b)`
  `a < b`          `a.__lt__(b)`
  `obj[i]`         `obj.__getitem__(i)`
  `obj()`          `obj.__call__()`
  `for x in obj`   `obj.__iter__()`

The goal is to make custom classes behave like Python's built-in types.

------------------------------------------------------------------------

# **str**()

## Purpose

Returns a human-readable representation of an object.

``` python
class Book:
    def __init__(self,title):
        self.title = title

    def __str__(self):
        return f"Book: {self.title}"
```

``` python
book = Book("Python")
print(book)
```

Output

``` text
Book: Python
```

Without `__str__`, Python prints a memory-oriented representation.

### Rules

-   Must return a string.
-   Intended for end users.
-   Normally used through `print(obj)` instead of calling directly.

------------------------------------------------------------------------

# **repr**()

## Purpose

Returns an unambiguous developer-oriented representation.

``` python
class Student:
    def __init__(self,name):
        self.name = name

    def __repr__(self):
        return f"Student('{self.name}')"
```

``` python
repr(student)
```

Output

``` text
Student('Alice')
```

### Difference

  **str**           **repr**
  ----------------- -----------------------------------------
  User-friendly     Developer-friendly
  Pretty output     Debugging output
  Used by print()   Used by repr(), containers and debugger

If `__str__` is missing, `print()` falls back to `__repr__`.

------------------------------------------------------------------------

# **len**()

Supports the built-in `len()`.

``` python
class Playlist:

    def __init__(self):
        self.songs=["A","B","C"]

    def __len__(self):
        return len(self.songs)
```

``` python
len(playlist)
```

Internally:

``` python
playlist.__len__()
```

Best practice: return a non-negative integer.

------------------------------------------------------------------------

# **add**()

Allows custom behavior for the `+` operator.

``` python
class Vector:

    def __init__(self,x,y):
        self.x=x
        self.y=y

    def __add__(self,other):

        if not isinstance(other,Vector):
            return NotImplemented

        return Vector(
            self.x+other.x,
            self.y+other.y
        )
```

``` python
v3 = v1 + v2
```

Internally:

``` python
v1.__add__(v2)
```

### Best Practice

-   Return a new object.
-   Do not modify operands.
-   Return `NotImplemented` for unsupported types.

------------------------------------------------------------------------

# **eq**()

Defines equality.

``` python
class Student:

    def __init__(self,name):
        self.name=name

    def __eq__(self,other):

        if not isinstance(other,Student):
            return NotImplemented

        return self.name==other.name
```

``` python
s1 == s2
```

↓

``` python
s1.__eq__(s2)
```

Without `__eq__`, Python compares object identity.

------------------------------------------------------------------------

# Comparison Dunders

  Operator   Method
  ---------- ----------
  `<`        `__lt__`
  `<=`       `__le__`
  `>`        `__gt__`
  `>=`       `__ge__`

Example:

``` python
def __lt__(self,other):
    if not isinstance(other,Student):
        return NotImplemented
    return self.marks < other.marks
```

These methods should return `True`, `False` or `NotImplemented`.

------------------------------------------------------------------------

# **getitem**()

Makes objects indexable.

``` python
class Team:

    def __init__(self):
        self.players=["Alice","Bob","Charlie"]

    def __getitem__(self,index):
        return self.players[index]
```

Usage:

``` python
team[0]
team[-1]
team[:2]
```

Internally:

``` python
team.__getitem__(0)
```

Python passes either an integer or a `slice` object.

------------------------------------------------------------------------

# **call**()

Makes an object callable like a function.

``` python
class Power:

    def __init__(self,power):
        self.power=power

    def __call__(self,x):
        return x**self.power
```

``` python
square = Power(2)

square(5)
```

Internally:

``` python
square.__call__(5)
```

Useful in AI/ML frameworks where models behave like functions.

------------------------------------------------------------------------

# **iter**()

Makes objects iterable.

``` python
class Team:

    def __init__(self):
        self.players=["Alice","Bob","Charlie"]

    def __iter__(self):
        return iter(self.players)
```

Usage:

``` python
for player in team:
    print(player)
```

Internally Python performs approximately:

``` python
iterator = team.__iter__()

while True:
    try:
        item = next(iterator)
    except StopIteration:
        break
```

Return an iterator, not the underlying list.

------------------------------------------------------------------------

# Best Practices

-   Implement only methods that make sense.
-   Return `NotImplemented` for unsupported operand types.
-   Keep behavior consistent with built-in types.
-   `__str__` → users.
-   `__repr__` → developers.
-   `+` should usually create a new object.
-   Avoid calling dunder methods directly in normal code.

------------------------------------------------------------------------

# Common Mistakes

-   Returning non-string from `__str__`.
-   Returning `None` instead of `NotImplemented`.
-   Mutating operands inside `__add__`.
-   Returning an iterable instead of an iterator from `__iter__`.
-   Confusing `__repr__` with `__str__`.

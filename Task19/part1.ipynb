# 1. Variables in OOP

Python has three major variable categories in classes.

  Variable Type   Belongs To   Lifetime                    Shared
  --------------- ------------ --------------------------- --------
  Local           Function     During function execution   ❌
  Instance        Object       Object lifetime             ❌
  Class           Class        Class lifetime              ✅

# 2. Local Variables

## Definition

Local variables exist only inside a function or method. They are created
when the function starts and destroyed when it finishes.

### Example

``` python
class Calculator:
    def add(self, a, b):
        result = a + b
        return result
```

-   `result` is a local variable.
-   It cannot be accessed outside `add()`.

### Lifetime

    Function Starts
          │
    Create Local Variables
          │
    Execute Code
          │
    Return
          │
    Destroy Local Variables

### Common Mistakes

``` python
class Demo:
    def show(self):
        x = 10

d = Demo()
print(d.x)   # AttributeError
```

Why?

Because `x` is **not** stored inside the object.

------------------------------------------------------------------------

# 3. Instance Variables

## Definition

Instance variables belong to individual objects and are created using
`self.variable`.

``` python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
```

Each object has its own copy.

### Memory Diagram

    s1
    ├── name = Alice
    └── age = 20

    s2
    ├── name = Bob
    └── age = 22

Changing one object does not affect another.

### Best Practices

-   Store object-specific state.
-   Initialize inside `__init__`.
-   Access using `self`.

------------------------------------------------------------------------

# 4. Class Variables

## Definition

Class variables belong to the class and are shared by every instance.

``` python
class Student:
    school = "ABC School"
```

### Memory

    Student
    └── school = "ABC School"

    s1
    └── name = Alice

    s2
    └── name = Bob

### Shadowing

``` python
s1.school = "XYZ"
```

This creates an **instance variable** named `school` instead of
modifying the class variable.

Modify shared data like this:

``` python
Student.school = "XYZ"
```

# 5. Instance Methods

Instance methods receive `self`.

``` python
class Student:

    def __init__(self, name):
        self.name = name

    def introduce(self):
        print(self.name)
```

Internal call:

``` python
student.introduce()
# approximately
Student.introduce(student)
```

Use when working with object-specific data.

------------------------------------------------------------------------

# 6. Class Methods

Decorator:

``` python
@classmethod
```

First parameter:

``` python
cls
```

Example:

``` python
class Student:

    school = "ABC"

    @classmethod
    def change_school(cls, name):
        cls.school = name
```

## Factory Methods

``` python
@classmethod
def from_string(cls, data):
    name, age = data.split(",")
    return cls(name, int(age))
```

Factory methods are alternative constructors.

------------------------------------------------------------------------

# 7. Static Methods

Decorator:

``` python
@staticmethod
```

No `self`.

No `cls`.

``` python
class Math:

    @staticmethod
    def square(x):
        return x * x
```

Used for helper functions logically related to the class.

------------------------------------------------------------------------

# Comparison

  Feature                Instance          Class            Static
  ---------------------- ----------------- ---------------- -----------
  Receives               `self`            `cls`            None
  Access Instance Data   ✅                ❌               ❌
  Access Class Data      ✅                ✅               ❌
  Typical Use            Object behavior   Class behavior   Utilities

------------------------------------------------------------------------

# Best Practices

-   Use local variables for temporary values.
-   Use instance variables for object state.
-   Use class variables for shared state.
-   Prefer class methods for factory constructors.
-   Prefer static methods for helper utilities.

------------------------------------------------------------------------

# Common Interview Questions

### Difference between local, instance and class variables?

-   Local → function only
-   Instance → one copy per object
-   Class → one shared copy

### Difference between instance, class and static methods?

-   Instance → object behavior
-   Class → class behavior
-   Static → helper logic

  Need              Use
  ----------------- -------------------

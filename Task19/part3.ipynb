# 1. What is Operator Overloading?

Operator overloading means **defining the behavior of Python operators
for your own classes** using special (dunder) methods.

Example:

``` python
v3 = v1 + v2
```

Internally:

``` python
v3 = v1.__add__(v2)
```

The same operator can perform different tasks depending on the object's
implementation.

------------------------------------------------------------------------

  ----------------- ----------------------
  `5 + 3`           Integer addition
  `"Py" + "thon"`   String concatenation
  `[1,2] + [3]`     List concatenation

The object decides what `+` means.

# 3. Overloading Arithmetic Operators

## Addition

``` python
class Vector:

    def __init__(self,x,y):
        self.x=x
        self.y=y

    def __add__(self,other):
        if not isinstance(other,Vector):
            return NotImplemented

        return Vector(
            self.x+other.x,
            self.y+other.y
        )
```

Usage

``` python
v3 = v1 + v2
```

------------------------------------------------------------------------

## Other Arithmetic Dunders

  Operator   Method
  ---------- ----------------
  `+`        `__add__`
  `-`        `__sub__`
  `*`        `__mul__`
  `/`        `__truediv__`
  `//`       `__floordiv__`
  `%`        `__mod__`
  `**`       `__pow__`

The implementation pattern is the same as `__add__`.

# 4. In-place Operators

Operator:

``` python
a += b
```

uses

``` python
__iadd__()
```

Example

``` python
def __iadd__(self,other):
    self.value += other.value
    return self
```

Unlike `__add__`, this usually modifies the existing object.

------------------------------------------------------------------------

# 5. Comparison Operators

Already covered:

-   `__eq__`
-   `__lt__`
-   `__le__`
-   `__gt__`
-   `__ge__`

These allow objects to work naturally with sorting and comparisons.

------------------------------------------------------------------------

# 6. Why Return NotImplemented?

Never do this:

``` python
return None
```

Instead:

``` python
return NotImplemented
```

This tells Python:

> "I don't know how to perform this operation."

Python can then try the reverse operation before raising `TypeError`.

------------------------------------------------------------------------

# 7. Design Principles

## Keep Operators Intuitive

Good

``` python
wallet3 = wallet1 + wallet2
```

Bad

``` python
wallet1 + wallet2
# unexpectedly deletes wallet2
```

Operator behavior should match user expectations.

------------------------------------------------------------------------

## Return New Objects

For arithmetic operators (`+`, `-`, `*`), prefer returning a new object.

Modify objects only for in-place operators such as `+=`.

------------------------------------------------------------------------

## Validate Operand Types

``` python
if not isinstance(other, Vector):
    return NotImplemented
```

Never assume the second operand is the correct type.

# 9. Common Mistakes

❌ Returning wrong types.

❌ Forgetting `NotImplemented`.

❌ Modifying operands unexpectedly.

❌ Making operator behavior confusing.

❌ Implementing operators that have no logical meaning.

------------------------------------------------------------------------

# 10. Best Practices

-   Make your classes behave like built-in types.
-   Keep operator behavior predictable.
-   Prefer immutable arithmetic operations.
-   Document custom behavior.
-   Use dunder methods only when they improve readability.

# 11. Interview Questions

### What is operator overloading?

The ability to customize Python operators for user-defined classes using
dunder methods.

------------------------------------------------------------------------

### Which dunder method supports `+`?

`__add__()`

------------------------------------------------------------------------

### Difference between `__add__()` and `__iadd__()`?

`__add__()` usually creates a new object.

`__iadd__()` usually modifies the current object (`+=`).

------------------------------------------------------------------------

### Why return `NotImplemented`?

It lets Python try the reverse operation before failing.

------------------------------------------------------------------------

### Should every class overload operators?

No.

Only overload operators that naturally match the meaning of your class.

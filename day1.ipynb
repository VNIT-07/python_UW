## I. What is Python?

### 1. Python is a **high-level**, **general-purpose programming language** that emphasizes code readability, simplicity, and ease-of-writing with the use of significant indentation, standard library, and garbage collection. 

#### - **High-level**: A programming language with strong **abstraction** (It hides the details) from the details of the computer.

#### - **General-Purpose**: Python is a general-purpose programming language, meaning it is designed to be **used for writing software** in a wide variety of application domains rather than being restricted to a single specialized task. 

#### **List-of-example:** web development, data science, machine learning, and system scripting. 

## II. Why it's most popular language and also mostly used for AI and ML? 

### **1. Why it's popular**

- Easy to Learn – Its simple syntax is similar to English, making it beginner-friendly.

- Easy to Read – Clean and readable code is easier to understand and maintain.

- Versatile – It can be used for web development, AI, machine learning, data science, automation, game development, and more.

- Large Community – Millions of developers use Python, so there are plenty of tutorials and support available.

- Rich Libraries – Thousands of built-in and third-party libraries save time and effort.

- Cross-Platform – Python programs run on Windows, macOS, and Linux with little or no changes.

- High Demand – Many companies use Python, making it a valuable skill for jobs and careers.

- Open Source – It is free to use and has continuous improvements from the global community.


### **2. Why use for AI and ML?**
- Artificial Intelligence (AI) and Machine Learning, Python is the preferred choice because it offers specialized, pre-built frameworks like TensorFlow, PyTorch, and Scikit-learn that accelerate model development and prototyping. 

## III. Difference between Python2 and Python3?

| Python 2                                | Python 3                                                    |
| --------------------------------------- | ----------------------------------------------------------- |
| Released in 2000                        | Released in 2008                                            |
| No longer supported (ended in 2020)     | Actively supported and updated                              |
| `print` is a statement: `print "Hello"` | `print` is a function: `print("Hello")`                     |
| `input()` behaves differently           | `input()` always returns a string                           |
| Default strings are ASCII               | Default strings are Unicode (better for multiple languages) |
| Integer division: `5/2` gives `2`       | Integer division: `5/2` gives `2.5`                         |
| Less secure and fewer features          | More secure, faster, and has many new features              |

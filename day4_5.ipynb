## Variables

A **variable** is a name that stores a value.

Python is **dynamically typed**, meaning you don't need to declare the variable's type. Python determines it automatically.


````python
name = "Vinit"
age = 22

print(name)
print(age)
print(type(name))
print(type(age))
````

## Dynamic Typing

The same variable can store different types of values during program execution.


````python
x = 10
print(type(x))

x = 3.14
print(type(x))

x = True
print(type(x))

x = "Python"
print(type(x))
````

# Built-in Data Types

## 1. Numbers
- `int`
- `float`
- `complex`


````python
integer = 25
floating = 3.14
complex_num = 2 + 5j

print(integer, type(integer))
print(floating, type(floating))
print(complex_num, type(complex_num))
````

## 2. Strings

````python
text = "Hello Python"

print(text)
print(text[0])
print(text[-1])
print(len(text))
````

## 3. Lists (Ordered, Mutable, Allow Duplicates)

````python
fruits = ["Apple", "Banana", "Orange"]

fruits.append("Grapes")
fruits[0] = "Mango"

print(fruits)
````

## 4. Tuples (Ordered, Immutable)

````python
student = ("Vinit", 22, "India")
print(student)
print(student[0])
````

## 5. Dictionaries (Key-Value Pairs)

````python
student = {
    "name": "Vinit",
    "age": 22,
    "city": "Ahmedabad"
}

student["course"] = "Python"

print(student)
print(student["name"])
````

## 6. Sets (Unique Values Only)

````python
numbers = {1,2,2,3,3,4}
numbers.add(5)

print(numbers)
````

## 7. Boolean

````python
is_student = True
print(is_student)
print(type(is_student))
````

## 8. None

````python
x = None
print(x)
print(type(x))
````

# Variable Naming Rules

✅ Valid
- `student_name`
- `_marks`
- `total_price`

❌ Invalid
- `1name`
- `first name`
- `my-name`
- `class`

## PEP 8
- Variables: `snake_case`
- Constants: `UPPER_CASE`
- Classes: `PascalCase`


## Multiple Assignment

````python
a, b, c = 10, 20, 30
print(a, b, c)

x = y = z = 100
print(x, y, z)

a, b = b, a
print(a, b)
````

## Type Checking with `type()`

````python
print(type(10))
print(type(3.14))
print(type("Python"))
print(type([1,2,3]))
print(type((1,2,3)))
print(type({"a":1}))
print(type({1,2,3}))
print(type(True))
print(type(None))
````

### Key Takeaways
- Python uses dynamic typing.
- Lists are mutable; tuples are immutable.
- Dictionaries store key-value pairs.
- Sets remove duplicates automatically.
- Follow PEP 8 naming conventions.
- Use `type()` to inspect an object's type.

## Mutable vs Immutable Objects in Python

### Overview

In Python, every variable refers to an **object**. Objects are of two types:

- **Mutable** → Can be modified after creation.
- **Immutable** → Cannot be modified after creation. Any modification creates a **new object**.

---

## Mutable Objects

#### Characteristics
- Can be modified after creation.
- The object's identity (`id`) usually remains the same.

#### Example

```python
fruits = ["Apple", "Banana"]

print("Before:", fruits)

fruits.append("Orange")   # Modify the same list

print("After :", fruits)
```

**Output**

```text
Before: ['Apple', 'Banana']
After : ['Apple', 'Banana', 'Orange']
```

---

## Immutable Objects

#### Characteristics
- Cannot be modified after creation.
- Any modification creates a new object.

#### Example

```python
name = "Python"

print("Before:", name)

name = "Java"    # Creates a new string object

print("After :", name)
```

**Output**

```text
Before: Python
After : Java
```

---

## Difference Using `id()`

```python
# Mutable Object
numbers = [1, 2, 3]

print(id(numbers))

numbers.append(4)

print(id(numbers))   # Same ID


# Immutable Object
x = 10

print(id(x))

x = x + 1

print(id(x))         # Different ID
```

---

## Comparison

| Feature | Mutable | Immutable |
|---------|---------|-----------|
| Can be modified? | ✅ Yes | ❌ No |
| New object created on change? | ❌ Usually No | ✅ Yes |
| Examples | `list`, `dict`, `set` | `int`, `str`, `tuple` |

---

### Memory Trick

> **Mutable = Modify the existing object.**  
> **Immutable = Create a new object when changing the value.**

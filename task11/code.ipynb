1. The Tip Calculator
Write a function named calculate_tip that takes two arguments: bill_amount and tip_percent.

Goal: Calculate the tip amount and return it.
Test: Call the function with a $50 bill and 20% tip, store the result in a variable, and print that variable.


```python
bill_amount= float(input("Enter bill amount: "))
def tip_calc(tip_amout, tip_percent=20):
    tip_amout = bill_amount*(tip_percent/100)
    final_bill = bill_amount-tip_amout
    print("Final Bill: $",final_bill)
    return tip_amout
print("Your bill: $",bill_amount)
print("The tip amount is: $",tip_calc(bill_amount))
```

2. The Age Checker

Goal: If the age is 18 or older, return True. Otherwise, return False.

```python
age = int(input("Enter Age:"))
def is_adult(a):
    if a>=18:
        print("Welcome, you are eligible")
        return True
    else:
        print("Access Denied")
        return False

is_adult(age)
    
    
```

3. The Username Creator
Write a function named create_username that takes first_name and last_name.

Goal: Combine the names into a single lowercase string (e.g., "John" + "Doe" → "johndoe") and return it.
Test: Call the function, save the result to a variable called user_id, and print user_id.

```python

firstN = input("Enter First Name: ").lower()
lastName = input("Enter Last Name: ").lower()

def create_username(FN, LN):
        return FN.replace(" ", "") + LN.replace(" ", "")
u_id =create_username(firstN,lastName)
print(u_id)
    
```

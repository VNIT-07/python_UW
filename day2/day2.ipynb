### II. What is Python Interpreter & REPL

#### 1. A **Python interpreter** is a program that reads, translates, and executes Python code line by line.

#### **=> How it works.**

- You write Python code in a .py /.ipynb file.

- The Python interpreter reads the code.

- It converts the code into bytecode.

- The Python Virtual Machine (PVM) executes the bytecode.

- The program produces the output.

#### 2. REPL

- REPL stands for **Read, Evaluate, Print, Loop**, and it is an interactive programming environment that allows users to execute Python code line by line and receive immediate feedback.

- Unlike running a full script, the REPL reads your input, evaluates the command, prints the result to the console, and then loops back to wait for the next instruction.

### III. Intendation & Syntax rules

**1. Indentation**: Indentation refers to the whitespace (spaces or tabs) at the beginning of a code line, which Python uses to define code blocks and logical structure, replacing the curly braces {} found in many other languages.

**2. Syntax Rules**: Key Syntax Rules for Indentation:

- **Block Definition**: Every block (e.g., inside if, for, while, def, or class) must start with a colon : and contain at least one indented line.

- **Standard Spacing**: The PEP 8 style guide recommends using 4 spaces per indentation level. 

- **Consistency**: All statements within the same block must have the same indentation level. 

- **No Mixing**: Python 3 strictly prohibits mixing tabs and spaces in the same file, raising a TabError. 

- **Nesting**: Nested blocks increase indentation by another 4 spaces for each level of depth. 


### IV. Standard Code & pep8

#### **1. Standard Code Practices in Python**

Standard code practices are guidelines that help developers write code that is clean, readable, consistent, and easy to maintain. They are not strict language rules, but following them makes it easier for others to understand and collaborate on your code.

**Some common standard practices include:**

- Use meaningful variable and function names.

- Keep functions short and focused on one task.

- Write comments and docstrings where needed.

- Follow consistent formatting and indentation.

- Avoid repeating the same code (DRY – Don't Repeat Yourself).

- Handle errors appropriately using exceptions.

- Organize imports and project files logically.


#### **2.** pep8

PEP 8 (Python Enhancement Proposal 8) is the official style guide for Python code. It defines conventions for formatting code so that Python programs look consistent across projects.

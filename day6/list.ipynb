# List

## 1. Overview

A **List** is an **ordered**, **mutable**, and **dynamic** collection of objects. It is one of the most commonly used data structures in Python because it can store multiple values of different data types and can grow or shrink dynamically.

Lists are extensively used in:

- Data Processing
- Machine Learning
- Web Development
- Automation
- APIs
- Data Structures & Algorithms

## Internal Working

Python implements a list as a **Dynamic Array**.

```
Index

0    1    2    3

+----+----+----+----+
|10  |20  |30  |    |
+----+----+----+----+
```

Python allocates extra memory so that `append()` usually doesn't require allocating a new array every time.

When capacity is full:

```
Old Array
+----+----+----+
|10  |20  |30  |
+----+----+----+

        │

        ▼

New Larger Array

+----+----+----+----+----+----+
|10  |20  |30  |40  |    |    |
+----+----+----+----+----+----+
```

This is why `append()` is **O(1) amortized** instead of always O(1).

# 6. List Methods

| Method | Purpose | Syntax | Example | Returns | Modifies Original | Complexity |
|---------|---------|--------|---------|----------|:-----------------:|------------|
| `append()` | Add one item | `lst.append(x)` | `[1,2].append(3)` → `[1,2,3]` | `None` | ✅ | O(1) amortized |
| `extend()` | Add iterable | `lst.extend(it)` | `[1,2].extend([3,4])` → `[1,2,3,4]` | `None` | ✅ | O(k) |
| `insert()` | Insert at index | `lst.insert(i,x)` | `[1,3].insert(1,2)` → `[1,2,3]` | `None` | ✅ | O(n) |
| `remove()` | Remove first matching value | `lst.remove(x)` | `[1,2,2].remove(2)` → `[1,2]` | `None` | ✅ | O(n) |
| `pop()` | Remove by index | `lst.pop(i)` | `[1,2,3].pop()` → `3` | Removed Item | ✅ | O(1)/O(n) |
| `clear()` | Remove all items | `lst.clear()` | `[1,2].clear()` → `[]` | `None` | ✅ | O(n) |
| `copy()` | Shallow copy | `lst.copy()` | `b=a.copy()` | `list` | ❌ | O(n) |
| `index()` | Find first occurrence | `lst.index(x)` | `[10,20].index(20)` → `1` | `int` | ❌ | O(n) |
| `count()` | Count occurrences | `lst.count(x)` | `[1,1,2].count(1)` → `2` | `int` | ❌ | O(n) |
| `reverse()` | Reverse list | `lst.reverse()` | `[1,2,3]` → `[3,2,1]` | `None` | ✅ | O(n) |
| `sort()` | Sort list | `lst.sort()` | `[3,1,2]` → `[1,2,3]` | `None` | ✅ | O(n log n) |
| `sorted()`* | Sorted copy | `sorted(lst)` | `sorted([3,1,2])` → `[1,2,3]` | `list` | ❌ | O(n log n) |

> **Note:** `sorted()` is a built-in function, not a list method.

# 7. Important Comparisons

## append() vs extend()

| append() | extend() |
|-----------|-----------|
| Adds one object | Adds every element |
| Can append a list as one item | Flattens one iterable level |
| `[1].append([2,3]) → [1,[2,3]]` | `[1].extend([2,3]) → [1,2,3]` |

---

## remove() vs pop()

| remove() | pop() |
|-----------|--------|
| Removes by value | Removes by index |
| Returns `None` | Returns removed element |
| Raises `ValueError` | Raises `IndexError` |

---

## sort() vs sorted()

| sort() | sorted() |
|----------|-----------|
| List method | Built-in function |
| Modifies original | Returns new list |
| Returns `None` | Returns sorted list |

---

## copy() vs Assignment

| `copy()` | `=` |
|-----------|------|
| Creates new outer list | Shares same object |
| Shallow copy | Reference copy |

---

# 9. Common Mistakes

### Mistake 1

```python
numbers = numbers.append(5)
```

❌ `append()` returns `None`.

---

### Mistake 2

```python
numbers.sort()

print(numbers.sort())
```

Returns `None`.

---

### Mistake 3

```python
a = [[0]*3]*3
```

All rows reference the same inner list.

Use:

```python
a = [[0]*3 for _ in range(3)]
```

---

### Mistake 4

```python
a.append([3,4])
```

does **not** behave like

```python
a.extend([3,4])
```

---

# 10. Best Practices

- Use `append()` for a single element.
- Use `extend()` for multiple elements.
- Use `sorted()` if you need to preserve the original list.
- Use `sort(key=...)` for custom sorting.
- Avoid modifying a list while iterating over it.
- Use `copy()` when you need a separate outer list.

# 13. Key Takeaways

- Lists are mutable dynamic arrays with fast indexing.
- `append()` is usually O(1) because Python overallocates memory.
- Choose `append()`, `extend()`, `sort()`, and `sorted()` based on whether you want to modify the original list.
- Understand shallow copies and reference behavior to avoid subtle bugs.
- Lists are one of the most fundamental data structures in Python and are used extensively in AI, data analysis, automation, and general software development.

````python
number = [2, 4, 6, 8, 10, 0, 1, 2, 3, 4, 5]
new_list = [20, 30, 40, 50, 60]

print("Original List:", number)
print("Length of List:", len(number))
print("Maximum Value:", max(number))
print("Minimum Value:", min(number))
print("Sum of Elements:", sum(number))

number.sort()
print("\nAfter sort():", number)

number.append(1000)
print("After append(1000):", number)

number.extend(new_list)
print("After extend(new_list):", number)

number.insert(10, 1099)
print("After insert(10, 1099):", number)

removed_item = number.pop(0)
print("Removed Item using pop(0):", removed_item)
print("After pop(0):", number)

index = number.index(10)
print("Index of 10:", index)

count = number.count(2)
print(count)

print("\nFinal List:", number)

# sorted() works on any iterable
text = "vinitpansuriya"
print("\nOriginal String:", text)
print("Sorted Characters:", sorted(text))
````

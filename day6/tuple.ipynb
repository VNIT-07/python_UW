## Creation Syntax

```python
t = (1,2,3)
t = 1,2,3
t = tuple(iterable)
empty = ()
single = (5,)
```

### Example

```python
t1 = (10, 20, 30)

t2 = 1, 2, 3

t3 = tuple([4, 5, 6])

empty = ()

single = (100,)

print(t1)
print(t2)
print(t3)
print(empty)
print(single)
```

**Use case:** Fixed collection of related values.

## Unpacking

```python

first,*middle,last = nums
```

### Example

```python

print(name)

print(course)

nums = (1, 2, 3, 4, 5)

first, *middle, last = nums

print(first)
print(middle)

```

## Built-in Functions

| Function | Syntax | Use Case |
|---|---|---|
| len | `len(t)` | Count elements |
| min | `min(t)` | Smallest element |
| max | `max(t)` | Largest element |
| sum | `sum(t)` | Sum numeric values |
| sorted | `sorted(t)` | Returns sorted **list** |
| any | `any(t)` | Any truthy value |
| all | `all(t)` | All truthy values |

### Example

```python
t = (5, 10, 15, 20)

print(len(t))

print(min(t))

print(max(t))

print(sum(t))

print(sorted(t))

print(any((0, False, 5)))

print(all((1, True, 10)))
```

## Hashability

Tuple is hashable **only if all elements are hashable**.

```python
hash((1,"AI"))
hash((1,[2]))   # TypeError
```

### Example

```python
t = (1, "Python")

print(hash(t))

# This raises an error because list is mutable
# print(hash((1, [2])))
```

Use Case:
- Dictionary keys
- Set elements

### Dictionary Key Example

```python
student = {
    (101, "Vinit"): "Python"
}

print(student[(101, "Vinit")])
```

## namedtuple

```python
from collections import namedtuple

Student = namedtuple("Student",["name","age"])
s = Student("Vatsal",21)
```

### Example

```python
from collections import namedtuple

Student = namedtuple("Student", ["name", "age", "course"])

s = Student("Vinit", 21, "Python")

print(s)

print(s.name)

print(s.age)

print(s.course)
```

Use case:
- Lightweight records
- Better readability than normal tuples

## Tuple vs List

| Feature | Tuple | List |
|---|---|---|
| Mutable | ❌ | ✅ |
| Memory | Less | More |
| Speed | Slightly Faster | Slightly Slower |
| Dict Key | Yes* | No |

### Example

```python
my_tuple = (1, 2, 3)

my_list = [1, 2, 3]

# Tuple cannot be modified
# my_tuple[0] = 100

my_list[0] = 100

print(my_tuple)

print(my_list)
```

## Time Complexity

| Operation | Complexity |
|---|---:|
| Index | O(1) |
| Slice | O(k) |
| Membership | O(n) |
| Iteration | O(n) |
| Concatenation | O(n+m) |

### Example

```python
t = (10, 20, 30, 40, 50)

# Index
print(t[2])

# Slice
print(t[1:4])

# Membership
print(30 in t)

# Iteration
for item in t:
    print(item)

# Concatenation
t2 = (60, 70)

print(t + t2)
```

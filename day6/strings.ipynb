# 1. String

### A string (str) in Python is a sequence of Unicode characters enclosed in single quotes (' '), double quotes (" "), or triple quotes (''' ''' or """ """).


## Overview
- Strings are **immutable Unicode sequences**.
- Almost all string methods return a **new string**.
- Prefer **f-strings** for formatting.
- Prefer **`"separator".join()`** over repeated `+=` for concatenation.

## String Methods Cheat Sheet

| Method | Purpose | Returns | Time | Example Use Case |
|--------|---------|---------|------|------------------|
| `upper()` | Convert to uppercase | New string | O(n) | `"hello".upper()` → `"HELLO"` |
| `lower()` | Convert to lowercase | New string | O(n) | `"HELLO".lower()` → `"hello"` |
| `capitalize()` | Capitalize first letter | New string | O(n) | `"python".capitalize()` → `"Python"` |
| `title()` | Convert to title case | New string | O(n) | `"hello world".title()` |
| `swapcase()` | Reverse letter case | New string | O(n) | `"PyThOn".swapcase()` |
| `casefold()` | Unicode-safe lowercase | New string | O(n) | Case-insensitive comparison |
| `find()` | Find substring | Index / -1 | O(n) | `"apple".find("p")` |
| `rfind()` | Find from right | Index / -1 | O(n) | Find last occurrence |
| `index()` | Find substring | Index / Error | O(n) | Raises error if not found |
| `rindex()` | Find from right | Index / Error | O(n) | Last occurrence with error |
| `count()` | Count occurrences | `int` | O(n) | `"banana".count("a")` |
| `replace()` | Replace text | New string | O(n) | `"cat".replace("c","b")` |
| `strip()` | Remove spaces both sides | New string | O(n) | `" hi ".strip()` |
| `lstrip()` | Remove left spaces | New string | O(n) | `" hi".lstrip()` |
| `rstrip()` | Remove right spaces | New string | O(n) | `"hi ".rstrip()` |
| `split()` | Split into list | `list` | O(n) | `"a,b".split(",")` |
| `rsplit()` | Split from right | `list` | O(n) | Split last delimiter |
| `splitlines()` | Split by lines | `list` | O(n) | Multi-line text |
| `partition()` | Split once | `tuple` | O(n) | `"a=b".partition("=")` |
| `join()` | Join iterable | New string | O(total chars) | `",".join(["a","b"])` |
| `center()` | Center text | New string | O(n) | `"Hi".center(10)` |
| `ljust()` | Left align | New string | O(n) | `"Hi".ljust(10)` |
| `rjust()` | Right align | New string | O(n) | `"Hi".rjust(10)` |
| `zfill()` | Pad with zeros | New string | O(n) | `"42".zfill(5)` |
| `format()` | Format string | New string | O(n) | `"{} {}".format("Hi","All")` |
| `format_map()` | Format using dictionary | New string | O(n) | `"{name}".format_map({"name":"Sam"})` |
| `encode()` | Convert to bytes | `bytes` | O(n) | `"Hi".encode()` |
| `translate()` | Character mapping | New string | O(n) | Replace multiple chars |

## Quick Notes
- **Immutable:** Original string never changes.
- **Return Value:** Most methods return a new string.
- **Searching:** `find()` returns `-1`; `index()` raises an error.
- **Whitespace:** `strip()` removes leading/trailing spaces.
- **Joining:** Use `join()` for efficient concatenation.
- **Formatting:** Prefer **f-strings** over `format()`.

## Delete

```python

```

### Example

```python

age = student.pop("age")

student.pop("course", "Key Not Found")

student.popitem()

print(student)
```

## Common Methods

| Method | Syntax | Use Case |
|---|---|---|
| get | `d.get(k,default)` | Safe lookup |
| setdefault | `d.setdefault(k,v)` | Initialize missing key |
| update | `d.update(x)` | Merge/update |
| pop | `d.pop(k)` | Remove & return value |
| popitem | `d.popitem()` | Remove last inserted |
| clear | `d.clear()` | Empty dictionary |
| copy | `d.copy()` | Shallow copy |

### Example

```python
student = {
    "name":"Vinit",
    "age":21
}

print(student.get("name"))

print(student.setdefault("city","Ahmedabad"))

student.update({"age":22})

print(student.pop("age"))

student.popitem()

copy_dict = student.copy()

student.clear()

print(copy_dict)
```

## Dictionary Views

| Method | Returns | Use Case |
|---|---|---|
| keys | dict_keys | Iterate keys |
| values | dict_values | Iterate values |
| items | dict_items | Key-value iteration |

Views are **live**, not copies.

### Example

```python
student = {
    "name":"Vinit",
    "age":21,
    "city":"Ahmedabad"
}

print(student.keys())

print(student.values())

print(student.items())

for key, value in student.items():
    print(key, value)
```

## Shallow Copy

```python
copy = d.copy()
```

Outer dictionary is copied.

Nested mutable objects remain shared.

### Example

```python
student = {
    "name":"Vinit",
    "marks":[90,95]
}

copy_student = student.copy()

copy_student["marks"].append(100)

print(student)
print(copy_student)
```

## Dictionary Comprehension

```python
{x:x*x for x in range(5)}

{k:v for k,v in d.items()}

{x:x*x for x in range(10) if x%2==0}
```

### Example

```python
square = {x:x*x for x in range(5)}

print(square)

student = {
    "a":90,
    "b":80,
    "c":70
}

new_dict = {k:v for k,v in student.items()}

print(new_dict)

even_square = {x:x*x for x in range(10) if x%2==0}

print(even_square)
```

Use cases:
- Build lookup tables
- Reverse mappings
- Data transformation

## Internal Working

Dictionary uses a **hash table**.

```
Key
 ↓
hash()
 ↓
Bucket
 ↓
Value
```

Average lookup: **O(1)**

### Example

```python
student = {
    "name":"Vinit",
    "age":21
}

print(hash("name"))

print(student["name"])
```

---

## Time Complexity

| Operation | Complexity |
|---|---:|
| Lookup | O(1) |
| Insert | O(1) |
| Update | O(1) |
| Delete | O(1) |
| Copy | O(n) |
| Iteration | O(n) |

### Example

```python
student = {
    "name":"Vinit",
    "age":21
}

# Lookup
print(student["name"])

# Insert
student["city"] = "Ahmedabad"

# Update
student["age"] = 22

# Delete
student.pop("city")

# Copy
copy_student = student.copy()

# Iteration
for key, value in student.items():
    print(key, value)
```

## Common Mistakes

- Mutable dictionary keys
- Confusing assignment with copy
- Assuming views are lists
- Forgetting duplicate keys overwrite previous values

### Example

```python
# Mutable key (Error)
# d = {[1,2]:100}

# Assignment vs Copy
a = {"x":1}
b = a

b["x"] = 100

print(a)

# Duplicate Keys
d = {
    "a":1,
    "a":10
}

print(d)
```

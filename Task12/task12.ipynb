# Modules

A **module** is a reusable Python `.py` file.

``` python
# calculator.py
def add(a,b):
    return a+b
```

Use:

``` python
import calculator
print(calculator.add(2,3))
```

## Import Styles

``` python
import math
from math import sqrt
from math import sqrt as square_root
```

Avoid:

``` python
from math import *
```

# **init**.py

Runs when the package is imported.

``` python
from .add import add
from .subtract import subtract

__version__="1.0.0"
```

Allows:

``` python
from math_pkg import add
```

# **name**

Direct execution:

``` python
print(__name__)
```

Output:

    __main__

Imported:

    module_name

# if **name** == "**main**"

``` python
def main():
    print("Program Started")

if __name__=="__main__":
    main()
```

This prevents executable code from running when imported.

# Virtual Environment

Create:

``` bash
python -m venv .venv
```

Activate (Windows):

``` bash
.venv\Scripts\activate
```

Install:

``` bash
pip install numpy
```

Freeze:

``` bash
pip freeze > requirements.txt
```

Install:

``` bash
pip install -r requirements.txt
```

Deactivate:

``` bash
deactivate
```

# Best Practices

-   One module, one responsibility.
-   Avoid wildcard imports.
-   Use `main()` with `if __name__ == "__main__":`.
-   Use one virtual environment per project.
-   Commit `requirements.txt`, not `.venv`.

# Summary

Modules organize reusable code. Packages organize related modules.
`__name__` identifies how a module is executed.
`if __name__ == "__main__":` protects executable code during imports.
`__init__.py` initializes packages and exports APIs. Virtual
environments isolate dependencies.

# DAY-15 Exception Handling

------------------------------------------------------------------------

# 1. Introduction

Exception handling allows a program to respond gracefully to runtime
errors instead of crashing. It separates **normal program logic** from
**error-handling logic**, making code more reliable and maintainable.

Common runtime exceptions include:

-   `ValueError`
-   `TypeError`
-   `ZeroDivisionError`
-   `IndexError`
-   `KeyError`
-   `FileNotFoundError`

------------------------------------------------------------------------

# 2. Why Exception Handling?

Without exception handling:

``` python
num = int(input("Enter a number: "))
print(100 / num)
print("Done")
```

Input:

    0

Program crashes with:

    ZeroDivisionError

Using exception handling prevents abrupt termination and allows recovery
or graceful failure.

------------------------------------------------------------------------

# 3. Exception Handling Flow

``` text
try
 │
 ├── No Exception ─────► else (if present)
 │                         │
 │                         ▼
 └── Exception ─► matching except
                           │
                           ▼
                       finally
                           │
                           ▼
                      Continue Program
```

------------------------------------------------------------------------

# 4. `try`

The `try` block contains code that **may raise an exception**.

``` python
try:
    num = int(input())
    print(100 / num)
```

------------------------------------------------------------------------

# 5. `except`

Handles matching exceptions.

``` python
try:
    num = int(input())
    print(100 / num)

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

## Multiple `except`

``` python
try:
    value = int(input())
    print(100 / value)

except ValueError:
    print("Enter a valid integer.")

except ZeroDivisionError:
    print("Division by zero is not allowed.")
```

## Catch Multiple Exceptions Together

``` python
except (ValueError, ZeroDivisionError):
    print("Invalid input.")
```

### Rules

-   Python checks `except` blocks from top to bottom.
-   Only the **first matching** handler executes.
-   Place **specific exceptions first** and general ones later.

------------------------------------------------------------------------

# 6. `else`

Runs **only if no exception occurs**.

``` python
try:
    result = 100 / 5

except ZeroDivisionError:
    print("Division error")

else:
    print(result)
```

Output

    20.0

------------------------------------------------------------------------

# 7. `finally`

Runs **whether an exception occurs or not**.

``` python
try:
    file = open("data.txt")
    data = file.read()

finally:
    file.close()
```

Typical uses:

-   Close files
-   Close database connections
-   Release locks
-   Clean temporary resources

### Important

Avoid returning from `finally` because it can suppress exceptions.

------------------------------------------------------------------------

# 8. `raise`

Used to manually create an exception.

``` python
age = int(input())

if age < 18:
    raise ValueError("Age must be at least 18.")
```

## Re-raising

``` python
try:
    print(10 / 0)

except ZeroDivisionError:
    print("Logging...")
    raise
```

------------------------------------------------------------------------

# 9. Custom Exceptions

Create exceptions specific to your application.

``` python
class AgeError(Exception):
    pass
```

Usage

``` python
if age < 18:
    raise AgeError("Minimum age is 18.")
```

### Constructor Example

``` python
class AgeError(Exception):

    def __init__(self, age):
        super().__init__(f"Invalid age: {age}")
```

------------------------------------------------------------------------

# 10. Common Built-in Exceptions

  Exception             Cause                         Example
  --------------------- ----------------------------- -----------------
  ValueError            Correct type, invalid value   `int("abc")`
  TypeError             Wrong type                    `"2" + 2`
  ZeroDivisionError     Divide by zero                `10/0`
  IndexError            Invalid list index            `lst[5]`
  KeyError              Missing dictionary key        `d["x"]`
  FileNotFoundError     Missing file                  `open("x.txt")`
  AttributeError        Missing attribute             `obj.x`
  ModuleNotFoundError   Missing module                `import xyz`

------------------------------------------------------------------------

# 11. Best Practices

-   Keep `try` blocks as small as practical.
-   Catch the most specific exception possible.
-   Use `else` for code that should execute only after success.
-   Use `finally` for cleanup.
-   Raise meaningful exception types.
-   Write clear error messages.
-   Create custom exceptions for business/domain rules.
-   Do not silently ignore exceptions.

------------------------------------------------------------------------

# 12. Common Pitfalls

❌ Bare `except:`

``` python
except:
    pass
```

❌ Catching `Exception` everywhere.

❌ Large `try` blocks.

❌ Returning from `finally`.

❌ Using custom exceptions when a built-in one is sufficient.

## Overview
List comprehensions provide a concise way to create new lists by combining iteration, transformation and optional filtering.

### Basic Syntax

```python
[expression for item in iterable]
```

```python
[expression for item in iterable if condition]
```

```python
[value_if_true if condition else value_if_false for item in iterable]
```

## Internal Working

```python
[x*x for x in nums]
```

Equivalent to

```python
result = []
for x in nums:
    result.append(x*x)
```

## Filtering

```python
[x for x in nums if x % 2 == 0]
```

## Inline if-else

```python
[x if x > 0 else 0 for x in nums]
```

Filtering removes elements.

Inline if-else transforms every element.

## Nested Loops

```python
[(i,j) for i in A for j in B]
```

Nested with filter

```python
[(i,j) for i in A for j in B if i != j]
```

## Nested Comprehensions

Flatten

```python
[item for row in matrix for item in row]
```

Transpose

```python
[[row[i] for row in matrix] for i in range(len(matrix[0]))]
```

## Generator Expressions

```python
(x*x for x in nums)
```

- Lazy evaluation
- One value at a time
- Low memory
- No indexing
- One-time iteration

## Scope

Python 3 keeps comprehension variables local.

## map() vs filter()

Prefer

```python
[x*x for x in nums]
[x for x in nums if x%2==0]
```

instead of lambda-based map/filter when readability is better.

## Time Complexity

| Operation | Complexity |
|---|---:|
| List Comprehension | O(n) |
| Generator Iteration | O(n) |
| Filter | O(n) |
| Nested Loops | O(n×m) |

## Common Mistakes

- Missing expression
- Confusing filtering and transformation
- Using comprehensions for side effects
- Expecting generators to support indexing
- Reusing exhausted generators

## Best Practices

- Use comprehensions for simple transformations.
- Use generators for huge datasets.
- Avoid deeply nested comprehensions.

## Interview Questions

## Cheat Sheet

```python
[x for x in nums]
[x*x for x in nums]
[x for x in nums if x>0]
[x if x>0 else 0 for x in nums]
[(i,j) for i in A for j in B]
(item for item in nums)
```

## Key Takeaways

- Comprehensions build new collections.
- Generators produce values lazily.
- Choose the right tool based on memory and readability.


````python
nums = [1,2,3,4]

result = []

for x in nums:
    result.append(x * x)

print(result)
````

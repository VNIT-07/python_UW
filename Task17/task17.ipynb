# TASK - 16 – Functional Programming

## Overview
Functional Programming (FP) is a programming paradigm where functions are treated as first-class citizens. Python supports FP alongside procedural and object-oriented programming.

---

# 1. First-Class Functions
Functions are objects. They can be assigned, passed, returned, and stored.

```python
def greet():
    return "Hello"

f = greet
print(f())
```

---

# 2. Higher-Order Functions

A higher-order function accepts another function as an argument and/or returns one.

```python
def add(a,b):
    return a+b

def calculate(func,x,y):
    return func(x,y)

print(calculate(add,2,3))
```

```python
def outer():
    def inner():
        print("Hello")
    return inner

f = outer()
f()
```

---

# 3. Lambda Functions

```python
square = lambda x: x*x
print(square(5))
```

Use lambda for short one-line functions.

---

# 4. map()

Transforms every element.

```python
nums=[1,2,3,4]
print(list(map(lambda x:x*x, nums)))
```

Returns a lazy iterator.

Multiple iterables:

```python
a=[1,2,3]
b=[10,20,30]
print(list(map(lambda x,y:x+y,a,b)))
```

Equivalent list comprehension:

```python
[x*x for x in nums]
```

---

# 5. filter()

Keeps elements satisfying a condition.

```python
nums=[1,2,3,4,5,6]
print(list(filter(lambda x:x%2==0, nums)))
```

Truthy filtering:

```python
data=[0,"",None,"Python",True]
print(list(filter(None,data)))
```

Equivalent comprehension:

```python
[x for x in nums if x%2==0]
```

---

# 6. reduce()

```python
from functools import reduce

nums=[1,2,3,4]
print(reduce(lambda x,y:x+y, nums))
```

Execution:

1+2=3 → 3+3=6 → 6+4=10

Initial value:

```python
reduce(lambda x,y:x+y,[1,2,3],100)
```

Prefer built-ins like sum() or max() when appropriate.

---

# 7. Closures

```python
def outer():
    message="Hello"
    def inner():
        print(message)
    return inner

f=outer()
f()
```

Function factory:

```python
def multiplier(n):
    def multiply(x):
        return x*n
    return multiply

double=multiplier(2)
print(double(10))
```

Requirements:
- Outer function
- Inner function
- Captured outer variable
- Inner survives after outer returns

---

# 8. Closures and Decorators

```python
def timer(func):
    def wrapper(*args, **kwargs):
        print("Starting")
        result=func(*args, **kwargs)
        print("Finished")
        return result
    return wrapper
```

Decorators work because wrapper is a closure.

---

# map() vs filter() vs reduce()

| Function | Purpose | Returns |
|---|---|---|
| map() | Transform | Iterator |
| filter() | Select | Iterator |
| reduce() | Aggregate | Single Value |

---

# Best Practices

- Use lambda only for simple expressions.
- Prefer comprehensions for readability.
- Use reduce() only when accumulation isn't covered by built-ins.
- Avoid overly complex functional pipelines.

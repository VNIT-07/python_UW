```python
# 1. map()

nums = [1, 2, 3, 4, 5]
res = map(lambda x: x*2, nums)
print(list(res))

nums = ["apple", "banana", "mango"]
res = map(lambda n: n.upper(), nums)
print(list(res))

nums = ["cat", "elephant", "dog"]
res = map(lambda n: len(n), nums)
print(list(res))

nums = [1, 2, 3, 4, 5]
res = map(lambda x: x*x, nums)
print(list(res))

nums = [1, 2, 3, 4, 5]
res = map(lambda x: str(x), nums)
print(list(res))


# 2. filter

nums = [1, 2, 3, 4, 5, 6]
res = filter(lambda x: x%2==0, nums)
print(list(res))

nums = [1, -2, -3, 4, -5, 6]
res = filter(lambda x: x>=0, nums)
print(list(res))

nums = ["cat", "apple", "dog", "banana"]
res = filter(lambda x: len(x)>4, nums)
print(list(res))

nums = [5, 12, 8, 20, 3, 15]
res = filter(lambda x: x>10, nums)
print(list(res))

nums = ["Apple", "Banana", "Avocado", "Mango"]
res = filter(lambda x: x[0]=="A", nums)
print(list(res))

# reduce

from functools import reduce

nums = [1, 2, 3, 4, 5, 8]
res = reduce(lambda x,y: x+y, nums)
print(res)

nums = [2, 3, 4, 5]
res = reduce(lambda x,y: x*y, nums)
print(res)


```
